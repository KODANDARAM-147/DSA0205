```python
# ==================== GOOGLE COLAB VERSION ====================
# Run this entire cell in Google Colab

!pip install opencv-python matplotlib pillow scipy scikit-image -q

import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files
import io
from PIL import Image
from time import time
import math
from scipy import ndimage
from mpl_toolkits.mplot3d import Axes3D

# ==================== 1. AUGMENTED REALITY VS VIRTUAL REALITY ====================

def demonstrate_ar_vs_vr():
    """
    Demonstrate AR vs VR concepts with visualizations
    """
    print("\n" + "="*70)
    print("1. AUGMENTED REALITY VS VIRTUAL REALITY")
    print("="*70)

    results = []
    titles = []

    # Create visualization for AR vs VR
    ar_vr_img = np.zeros((500, 800, 3), dtype=np.uint8)

    # AR Section (Left half)
    ar_vr_img[0:500, 0:400, 0] = 50
    ar_vr_img[0:500, 0:400, 1] = 50
    ar_vr_img[0:500, 0:400, 2] = 200

    cv2.putText(ar_vr_img, "AUGMENTED REALITY (AR)", (20, 50), cv2.FONT_HERSHEY_SIMPLEX, 0.8, (255,255,255), 2)

    cv2.rectangle(ar_vr_img, (50, 100), (150, 200), (0, 255, 0), 2)
    cv2.putText(ar_vr_img, "Real World", (60, 90), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0,255,0), 1)
    cv2.putText(ar_vr_img, "+", (180, 150), cv2.FONT_HERSHEY_SIMPLEX, 1, (255,255,255), 2)
    cv2.rectangle(ar_vr_img, (200, 100), (300, 200), (0, 0, 255), 2)
    cv2.putText(ar_vr_img, "Virtual Object", (210, 90), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0,0,255), 1)

    cv2.rectangle(ar_vr_img, (100, 250), (300, 400), (255,255,255), 2)
    cv2.putText(ar_vr_img, "Smartphone/Glasses", (120, 240), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (255,255,255), 1)
    cv2.putText(ar_vr_img, "Overlay digital info on real world", (50, 450), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (255,255,255), 1)
    cv2.putText(ar_vr_img, "Real + Virtual = AR", (50, 480), cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0,255,255), 2)

    # VR Section (Right half)
    ar_vr_img[0:500, 400:800, 0] = 100
    ar_vr_img[0:500, 400:800, 1] = 0
    ar_vr_img[0:500, 400:800, 2] = 100

    cv2.putText(ar_vr_img, "VIRTUAL REALITY (VR)", (420, 50), cv2.FONT_HERSHEY_SIMPLEX, 0.8, (255,255,255), 2)

    cv2.rectangle(ar_vr_img, (450, 100), (750, 200), (255,255,255), 2)
    cv2.putText(ar_vr_img, "VR Headset", (480, 90), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (255,255,255), 1)

    cv2.circle(ar_vr_img, (600, 300), 50, (0, 255, 255), 2)
    cv2.circle(ar_vr_img, (500, 350), 30, (255, 0, 255), 2)
    cv2.circle(ar_vr_img, (700, 320), 40, (0, 255, 0), 2)
    cv2.putText(ar_vr_img, "Fully Virtual World", (500, 250), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (255,255,255), 1)
    cv2.putText(ar_vr_img, "Complete immersion in virtual environment", (430, 450), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (255,255,255), 1)
    cv2.putText(ar_vr_img, "Fully Virtual = VR", (450, 480), cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0,255,255), 2)

    results.append(ar_vr_img)
    titles.append("AR vs VR Concept")

    # Comparison Table
    comparison_img = np.zeros((400, 800, 3), dtype=np.uint8)
    comparison_img[:, :, :] = 20

    headers = ["Feature", "AR", "VR"]
    features = [
        ["Environment", "Real + Virtual", "Fully Virtual"],
        ["Device", "Phone/Glasses", "Headset"],
        ["Immersion", "Partial", "Full"],
        ["Interaction", "Real-time overlay", "Fully controlled"],
        ["Application", "Navigation, Games", "Gaming, Training"]
    ]

    for i, header in enumerate(headers):
        x = 50 + i * 250
        cv2.putText(comparison_img, header, (x, 50), cv2.FONT_HERSHEY_SIMPLEX, 0.8, (255,255,0), 2)

    for row, (feature, ar, vr) in enumerate(features):
        y = 100 + row * 55
        cv2.putText(comparison_img, feature, (50, y), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (255,255,255), 1)
        cv2.putText(comparison_img, ar, (250, y), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0,255,0), 1)
        cv2.putText(comparison_img, vr, (500, y), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0,255,255), 1)
        cv2.line(comparison_img, (0, y+5), (800, y+5), (100,100,100), 1)

    results.append(comparison_img)
    titles.append("AR vs VR Comparison")

    print("    ✓ AR vs VR concept visualized")
    print("    ✓ AR: Real world + Virtual overlay")
    print("    ✓ VR: Fully virtual environment")

    return results, titles

# ==================== 2. CAMERA MODEL AND CALIBRATION ====================

def demonstrate_camera_model_calibration(img):
    """
    Demonstrate Camera Model and Calibration:
    1. Pinhole Camera Model
    2. Camera Parameters
    3. Calibration Pattern Detection
    """
    print("\n" + "="*70)
    print("2. CAMERA MODEL AND CALIBRATION")
    print("="*70)

    print("\n[Mathematical Model]")
    print("Pinhole Camera Model:")
    print("[x']   [f  0  cx] [X]")
    print("[y'] = [0  f  cy] [Y]")
    print("[1 ]   [0  0  1 ] [Z]")
    print("Where (cx, cy) = principal point, f = focal length")

    results = []
    titles = []

    # Original
    results.append(img)
    titles.append("Original Image")

    # 1. Pinhole Camera Model Visualization
    print("\n[1] Pinhole Camera Model")
    camera_img = np.zeros((400, 600, 3), dtype=np.uint8)
    camera_img[:, :, :] = 30

    # Draw camera
    cv2.rectangle(camera_img, (150, 150), (200, 200), (255,255,255), 2)
    cv2.rectangle(camera_img, (100, 100), (150, 150), (255,255,255), 2)
    cv2.putText(camera_img, "Camera", (170, 140), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (255,255,255), 1)

    # Draw optical axis
    cv2.line(camera_img, (175, 175), (400, 175), (0,255,255), 2)
    cv2.putText(camera_img, "Optical Axis", (200, 170), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0,255,255), 1)

    # Draw focal length
    cv2.line(camera_img, (175, 175), (175, 250), (0,255,0), 2)
    cv2.putText(camera_img, "f (focal length)", (180, 240), cv2.FONT_HERSHEY_SIMPLEX, 0.4, (0,255,0), 1)

    # Draw image plane
    cv2.rectangle(camera_img, (400, 100), (450, 250), (255,0,255), 2)
    cv2.putText(camera_img, "Image Plane", (420, 90), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (255,0,255), 1)

    # Draw world point
    cv2.circle(camera_img, (50, 50), 5, (0,255,255), -1)
    cv2.putText(camera_img, "World Point (X,Y,Z)", (10, 40), cv2.FONT_HERSHEY_SIMPLEX, 0.4, (0,255,255), 1)
    cv2.line(camera_img, (50, 50), (175, 175), (255,255,255), 1)

    results.append(camera_img)
    titles.append("Pinhole Camera Model")
    print(f"    ✓ Pinhole camera model visualized")

    # 2. Camera Intrinsic Parameters
    print("\n[2] Camera Intrinsic Parameters")
    intrinsic_img = np.zeros((300, 500, 3), dtype=np.uint8)
    intrinsic_img[:, :, :] = 20

    params = [
        "Camera Intrinsic Parameters Matrix (K):",
        "",
        "[fx  0  cx]",
        "[0  fy  cy]",
        "[0   0   1 ]",
        "",
        "fx, fy = Focal lengths in pixels",
        "cx, cy = Principal point (image center)"
    ]

    for i, line in enumerate(params):
        cv2.putText(intrinsic_img, line, (20, 30 + i*30), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0,255,255), 1)

    results.append(intrinsic_img)
    titles.append("Camera Parameters")
    print(f"    ✓ Camera intrinsic parameters displayed")

    # 3. Camera Calibration Pattern (Checkerboard)
    print("\n[3] Camera Calibration Pattern")
    chessboard = create_chessboard_pattern(8, 6, 50)
    cv2.putText(chessboard, "Calibration Pattern (Checkerboard)", (10, 30), cv2.FONT_HERSHEY_SIMPLEX, 0.5, 255, 2)
    results.append(chessboard)
    titles.append("Calibration Pattern")
    print(f"    ✓ Chessboard calibration pattern created")

    # 4. Distortion Model
    print("\n[4] Lens Distortion Model")
    distortion_img = np.zeros((400, 600, 3), dtype=np.uint8)
    distortion_img[:, :, :] = 20

    distortion_text = [
        "Lens Distortion Models:",
        "",
        "Radial Distortion:",
        "x_corrected = x * (1 + k1*r² + k2*r⁴ + k3*r⁶)",
        "y_corrected = y * (1 + k1*r² + k2*r⁴ + k3*r⁶)",
        "",
        "Tangential Distortion:",
        "x_corrected = x + [2*p1*x*y + p2*(r²+2*x²)]",
        "y_corrected = y + [p1*(r²+2*y²) + 2*p2*x*y]",
        "",
        "Distortion Coefficients: [k1, k2, p1, p2, k3]"
    ]

    for i, line in enumerate(distortion_text):
        cv2.putText(distortion_img, line, (20, 30 + i*35), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0,255,255), 1)

    results.append(distortion_img)
    titles.append("Distortion Model")
    print(f"    ✓ Lens distortion model displayed")

    # 5. Camera Calibration Process
    print("\n[5] Camera Calibration Process")
    calibration_img = np.zeros((300, 600, 3), dtype=np.uint8)
    calibration_img[:, :, :] = 20

    calib_text = [
        "Camera Calibration Steps:",
        "",
        "1. Capture images of calibration pattern",
        "2. Detect corner points in pattern",
        "3. Establish 2D-3D correspondences",
        "4. Solve for camera matrix using PnP",
        "5. Estimate distortion coefficients",
        "6. Optimize with Levenberg-Marquardt"
    ]

    for i, line in enumerate(calib_text):
        cv2.putText(calibration_img, line, (20, 30 + i*32), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0,255,255), 1)

    results.append(calibration_img)
    titles.append("Calibration Process")
    print(f"    ✓ Camera calibration steps displayed")

    return results, titles

def create_chessboard_pattern(rows, cols, square_size):
    """Create a chessboard calibration pattern"""
    img = np.zeros((rows * square_size, cols * square_size), dtype=np.uint8)

    for i in range(rows):
        for j in range(cols):
            if (i + j) % 2 == 0:
                img[i*square_size:(i+1)*square_size, j*square_size:(j+1)*square_size] = 255

    return img

# ==================== 3. BINOCULAR IMAGING SYSTEM ====================

def demonstrate_binocular_system(img):
    """
    Demonstrate Binocular Imaging System:
    1. Stereo Camera Setup
    2. Epipolar Geometry
    3. Disparity and Depth
    4. 3D Reconstruction
    """
    print("\n" + "="*70)
    print("3. BINOCULAR IMAGING SYSTEM")
    print("="*70)

    print("\n[Mathematical Model]")
    print("Depth from Disparity:")
    print("Z = (f * B) / d")
    print("Where Z = depth, f = focal length, B = baseline, d = disparity")

    results = []
    titles = []

    # Original
    results.append(img)
    titles.append("Original Image")

    # 1. Stereo Camera Setup
    print("\n[1] Stereo Camera Setup")
    stereo_img = np.zeros((400, 600, 3), dtype=np.uint8)
    stereo_img[:, :, :] = 20

    # Draw two cameras
    cv2.rectangle(stereo_img, (100, 200), (150, 250), (255,0,0), 2)
    cv2.putText(stereo_img, "Left", (110, 195), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (255,0,0), 1)

    cv2.rectangle(stereo_img, (450, 200), (500, 250), (0,255,0), 2)
    cv2.putText(stereo_img, "Right", (460, 195), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0,255,0), 1)

    # Draw baseline
    cv2.line(stereo_img, (125, 225), (475, 225), (255,255,255), 2)
    cv2.putText(stereo_img, "Baseline B", (250, 220), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (255,255,255), 1)

    # Draw object
    cv2.circle(stereo_img, (300, 150), 10, (0,255,255), -1)
    cv2.putText(stereo_img, "Object", (290, 140), cv2.FONT_HERSHEY_SIMPLEX, 0.4, (0,255,255), 1)

    # Draw rays
    cv2.line(stereo_img, (125, 225), (300, 150), (255,255,0), 1)
    cv2.line(stereo_img, (475, 225), (300, 150), (255,255,0), 1)

    # Draw image planes
    cv2.rectangle(stereo_img, (50, 280), (200, 330), (255,0,255), 1)
    cv2.putText(stereo_img, "Left Image", (70, 275), cv2.FONT_HERSHEY_SIMPLEX, 0.4, (255,0,255), 1)

    cv2.rectangle(stereo_img, (400, 280), (550, 330), (255,0,255), 1)
    cv2.putText(stereo_img, "Right Image", (420, 275), cv2.FONT_HERSHEY_SIMPLEX, 0.4, (255,0,255), 1)

    # Draw epipolar lines
    cv2.line(stereo_img, (50, 300), (200, 300), (0,255,255), 1)
    cv2.line(stereo_img, (400, 300), (550, 300), (0,255,255), 1)
    cv2.putText(stereo_img, "Epipolar Lines", (220, 300), cv2.FONT_HERSHEY_SIMPLEX, 0.4, (0,255,255), 1)

    results.append(stereo_img)
    titles.append("Stereo Camera System")
    print(f"    ✓ Stereo camera setup visualized")

    # 2. Epipolar Geometry
    print("\n[2] Epipolar Geometry")
    epipolar_img = np.zeros((300, 600, 3), dtype=np.uint8)
    epipolar_img[:, :, :] = 20

    epipolar_text = [
        "Epipolar Geometry:",
        "",
        "• Epipole: Projection of one camera in the other",
        "• Epipolar Line: Line connecting epipole to point",
        "• Epipolar Plane: Plane through 3D point and cameras",
        "• Fundamental Matrix F: Maps points to epipolar lines",
        "",
        "x'^T * F * x = 0"
    ]

    for i, line in enumerate(epipolar_text):
        cv2.putText(epipolar_img, line, (20, 30 + i*35), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0,255,255), 1)

    results.append(epipolar_img)
    titles.append("Epipolar Geometry")
    print(f"    ✓ Epipolar geometry explained")

    # 3. Disparity Map Visualization - FIXED
    print("\n[3] Disparity Map")
    disp_map = np.zeros((200, 300), dtype=np.uint8)

    for i in range(200):
        for j in range(300):
            # Foreground (closer)
            dist = np.sqrt((i-100)**2 + (j-150)**2)
            if dist < 40:
                disp_map[i, j] = 200 - int(dist * 0.5)
            else:
                disp_map[i, j] = 50 + int((abs(i-100) * 0.2) + (abs(j-150) * 0.2))

    disp_map = np.clip(disp_map, 0, 255).astype(np.uint8)
    cv2.putText(disp_map, "Disparity Map (Brighter = Closer)", (10, 20), cv2.FONT_HERSHEY_SIMPLEX, 0.4, 255, 1)
    results.append(disp_map)
    titles.append("Disparity Map")
    print(f"    ✓ Synthetic disparity map created")

    # 4. Depth from Disparity - FIXED
    print("\n[4] Depth from Disparity")
    focal_length = 1000
    baseline = 200
    depth_map = np.zeros_like(disp_map, dtype=np.float32)

    for i in range(disp_map.shape[0]):
        for j in range(disp_map.shape[1]):
            disp_val = disp_map[i, j]
            if disp_val > 0:
                depth_val = (focal_length * baseline) / (disp_val + 1)
                depth_map[i, j] = depth_val

    # Handle invalid values
    valid_mask = depth_map > 0
    if np.any(valid_mask):
        depth_map_valid = depth_map[valid_mask]
        depth_min = np.min(depth_map_valid)
        depth_max = np.max(depth_map_valid)
        if depth_max > depth_min:
            depth_map[valid_mask] = ((depth_map[valid_mask] - depth_min) / (depth_max - depth_min)) * 255
        else:
            depth_map[valid_mask] = 128

    depth_map = depth_map.astype(np.uint8)
    cv2.putText(depth_map, "Depth Map (Brighter = Closer)", (10, 20), cv2.FONT_HERSHEY_SIMPLEX, 0.4, 255, 1)
    results.append(depth_map)
    titles.append("Depth Map")
    print(f"    ✓ Depth map computed from disparity")

    # 5. 3D Point Cloud
    print("\n[5] 3D Reconstruction from Stereo")
    point_cloud = np.zeros((400, 600, 3), dtype=np.uint8)
    point_cloud[:, :, :] = 20

    # Generate synthetic 3D points
    np.random.seed(42)
    points = []
    for i in range(100):
        x = np.random.uniform(50, 550)
        y = np.random.uniform(50, 350)
        z = np.random.uniform(50, 200)
        points.append([x, y, z])

    for p in points:
        scale = 500 / (p[2] + 100)
        px = int(300 + (p[0] - 300) * scale * 0.5)
        py = int(200 + (p[1] - 200) * scale * 0.5)
        if 0 <= px < 600 and 0 <= py < 400:
            size = int(3 * scale)
            color = (int(200 - p[2]), int(p[2]), 100)
            cv2.circle(point_cloud, (px, py), max(1, size), color, -1)

    cv2.putText(point_cloud, "3D Point Cloud Reconstruction", (10, 30), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (255,255,255), 1)
    results.append(point_cloud)
    titles.append("3D Point Cloud")
    print(f"    ✓ 3D point cloud visualization created")

    return results, titles

# ==================== 4. MULTIPLE VIEW GEOMETRY ====================

def demonstrate_multiple_view_geometry():
    """
    Demonstrate Multiple View Geometry:
    1. Fundamental Matrix
    2. Essential Matrix
    3. Homography
    4. Structure from Motion
    """
    print("\n" + "="*70)
    print("4. MULTIPLE VIEW GEOMETRY")
    print("="*70)

    results = []
    titles = []

    # 1. Fundamental Matrix
    print("\n[1] Fundamental Matrix")
    fmat_img = np.zeros((300, 600, 3), dtype=np.uint8)
    fmat_img[:, :, :] = 20

    fmat_text = [
        "Fundamental Matrix F:",
        "",
        "x'^T * F * x = 0",
        "",
        "• 3x3 matrix with rank 2",
        "• Maps points in one image to lines in the other",
        "• Encodes epipolar geometry",
        "• Computed from 8+ point correspondences",
        "",
        "Properties: det(F) = 0"
    ]

    for i, line in enumerate(fmat_text):
        cv2.putText(fmat_img, line, (20, 30 + i*30), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0,255,255), 1)

    results.append(fmat_img)
    titles.append("Fundamental Matrix")
    print(f"    ✓ Fundamental matrix explained")

    # 2. Essential Matrix
    print("\n[2] Essential Matrix")
    emat_img = np.zeros((300, 600, 3), dtype=np.uint8)
    emat_img[:, :, :] = 20

    emat_text = [
        "Essential Matrix E:",
        "",
        "E = [t]_x * R",
        "",
        "• 3x3 matrix with rank 2",
        "• Encodes rotation R and translation t",
        "• Used for calibrated cameras",
        "• E = K'^T * F * K",
        "",
        "Properties: det(E) = 0, 2E*E^T*E - trace(E*E^T)*E = 0"
    ]

    for i, line in enumerate(emat_text):
        cv2.putText(emat_img, line, (20, 30 + i*30), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0,255,255), 1)

    results.append(emat_img)
    titles.append("Essential Matrix")
    print(f"    ✓ Essential matrix explained")

    # 3. Homography
    print("\n[3] Homography")
    homography_img = np.zeros((300, 600, 3), dtype=np.uint8)
    homography_img[:, :, :] = 20

    homography_text = [
        "Homography H:",
        "",
        "x' = H * x",
        "",
        "• 3x3 invertible matrix",
        "• Maps points between planes",
        "• Used for panorama stitching",
        "• Computed from 4+ point correspondences",
        "",
        "Applications: Image stitching, perspective correction"
    ]

    for i, line in enumerate(homography_text):
        cv2.putText(homography_img, line, (20, 30 + i*30), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0,255,255), 1)

    results.append(homography_img)
    titles.append("Homography")
    print(f"    ✓ Homography explained")

    # 4. Structure from Motion Pipeline
    print("\n[4] Structure from Motion")
    sfm_img = np.zeros((350, 600, 3), dtype=np.uint8)
    sfm_img[:, :, :] = 20

    sfm_text = [
        "Structure from Motion (SfM) Pipeline:",
        "",
        "1. Feature Extraction (SIFT, ORB)",
        "2. Feature Matching across images",
        "3. Compute Fundamental/Essential Matrix",
        "4. Estimate Camera Motion (R, t)",
        "5. Triangulate 3D points",
        "6. Bundle Adjustment (Refine)",
        "7. Dense Reconstruction",
        "",
        "Output: 3D structure + Camera poses"
    ]

    for i, line in enumerate(sfm_text):
        cv2.putText(sfm_img, line, (20, 30 + i*30), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0,255,255), 1)

    results.append(sfm_img)
    titles.append("SfM Pipeline")
    print(f"    ✓ Structure from Motion pipeline explained")

    return results, titles

# ==================== 5. STRUCTURE DETERMINATION ====================

def demonstrate_structure_determination():
    """
    Demonstrate Structure Determination:
    1. Triangulation
    2. Structure from Motion
    3. Depth Estimation
    4. 3D Reconstruction
    """
    print("\n" + "="*70)
    print("5. STRUCTURE DETERMINATION")
    print("="*70)

    print("\n[Mathematical Model]")
    print("Triangulation:")
    print("X = (P1^T * x1) × (P2^T * x2)")
    print("Where P1, P2 are camera projection matrices")

    results = []
    titles = []

    # 1. Triangulation
    print("\n[1] Triangulation")
    tri_img = np.zeros((300, 600, 3), dtype=np.uint8)
    tri_img[:, :, :] = 20

    tri_text = [
        "Triangulation:",
        "",
        "Given: Corresponding points in two images",
        "P1 and P2 (camera projection matrices)",
        "",
        "Solution:",
        "X = argmin(||P1*X - x1||² + ||P2*X - x2||²)",
        "",
        "Methods:",
        "• Linear Triangulation",
        "• Non-linear optimization",
        "• Midpoint method"
    ]

    for i, line in enumerate(tri_text):
        cv2.putText(tri_img, line, (20, 30 + i*30), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0,255,255), 1)

    results.append(tri_img)
    titles.append("Triangulation")
    print(f"    ✓ Triangulation explained")

    # 2. 3D Reconstruction Visualization
    print("\n[2] 3D Reconstruction")
    recon_img = np.zeros((400, 600, 3), dtype=np.uint8)
    recon_img[:, :, :] = 20

    # Generate 3D points forming a cube
    cube_points = []
    for x in [-1, 1]:
        for y in [-1, 1]:
            for z in [-1, 1]:
                cube_points.append([x, y, z])

    def project_3d_to_2d(points, rot_angle=0.3):
        projected = []
        for p in points:
            x = p[0] * math.cos(rot_angle) - p[2] * math.sin(rot_angle)
            y = p[1]
            z = p[0] * math.sin(rot_angle) + p[2] * math.cos(rot_angle)
            f = 300
            px = 300 + int(f * x / (z + 3))
            py = 200 + int(f * y / (z + 3))
            projected.append([px, py])
        return projected

    proj_points = project_3d_to_2d(cube_points)

    edges = [(0,1), (0,2), (0,4), (1,3), (1,5), (2,3), (2,6), (3,7), (4,5), (4,6), (5,7), (6,7)]

    for edge in edges:
        pt1 = proj_points[edge[0]]
        pt2 = proj_points[edge[1]]
        if 0 <= pt1[0] < 600 and 0 <= pt1[1] < 400 and 0 <= pt2[0] < 600 and 0 <= pt2[1] < 400:
            cv2.line(recon_img, tuple(pt1), tuple(pt2), (0,255,255), 2)

    for pt in proj_points:
        if 0 <= pt[0] < 600 and 0 <= pt[1] < 400:
            cv2.circle(recon_img, tuple(pt), 4, (0,0,255), -1)

    cv2.putText(recon_img, "3D Reconstruction (Cube)", (10, 30), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (255,255,255), 1)
    results.append(recon_img)
    titles.append("3D Reconstruction")
    print(f"    ✓ 3D reconstruction visualization created")

    # 3. Depth Map from Structure
    print("\n[3] Depth Map from Structure")
    depth_img = np.zeros((200, 300), dtype=np.uint8)

    for i in range(200):
        for j in range(300):
            if (i-100)**2 + (j-150)**2 < 900:
                depth_img[i, j] = 200
            elif j < 150:
                depth_img[i, j] = int(100 + j * 0.5)
            else:
                depth_img[i, j] = int(50 + (300-j) * 0.5)

    depth_img = np.clip(depth_img, 0, 255).astype(np.uint8)
    cv2.putText(depth_img, "Depth Map (from Structure)", (10, 20), cv2.FONT_HERSHEY_SIMPLEX, 0.4, 255, 1)
    results.append(depth_img)
    titles.append("Depth Map")
    print(f"    ✓ Depth map from structure created")

    return results, titles

# ==================== 6. SHAPE FROM SHADING ====================

def demonstrate_shape_from_shading(img):
    """
    Demonstrate Shape from Shading:
    1. Lambertian Reflection Model
    2. Photometric Stereo
    3. Surface Normal Estimation
    4. 3D Shape Recovery
    """
    print("\n" + "="*70)
    print("6. SHAPE FROM SHADING")
    print("="*70)

    print("\n[Mathematical Model]")
    print("Lambertian Reflection:")
    print("I = ρ * L · N")
    print("Where I = intensity, ρ = albedo, L = light direction, N = surface normal")

    results = []
    titles = []

    # Original
    results.append(img)
    titles.append("Original Image")

    # 1. Lambertian Reflection Model
    print("\n[1] Lambertian Reflection Model")
    lambertian_img = np.zeros((300, 600, 3), dtype=np.uint8)
    lambertian_img[:, :, :] = 20

    lambertian_text = [
        "Lambertian Reflection Model:",
        "",
        "I(x,y) = ρ * (L · N)",
        "",
        "Where:",
        "• I = Pixel intensity",
        "• ρ = Albedo (surface reflectance)",
        "• L = Light direction vector",
        "• N = Surface normal vector",
        "",
        "Assumption: Surface is perfectly diffuse",
        "No specular reflections"
    ]

    for i, line in enumerate(lambertian_text):
        cv2.putText(lambertian_img, line, (20, 30 + i*30), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0,255,255), 1)

    results.append(lambertian_img)
    titles.append("Lambertian Model")
    print(f"    ✓ Lambertian reflection model explained")

    # 2. Surface Normal Visualization
    print("\n[2] Surface Normal Visualization")
    normal_img = np.zeros((300, 400, 3), dtype=np.uint8)
    normal_img[:, :, :] = 20

    for i in range(150):
        for j in range(150):
            x = (i - 75) / 75
            y = (j - 75) / 75
            if x*x + y*y < 1:
                z = np.sqrt(max(0, 1 - x*x - y*y))
                nx = x / max(0.01, np.sqrt(x*x + y*y + z*z))
                ny = y / max(0.01, np.sqrt(x*x + y*y + z*z))
                nz = z / max(0.01, np.sqrt(x*x + y*y + z*z))

                r = int((nx + 1) / 2 * 255)
                g = int((ny + 1) / 2 * 255)
                b = int((nz + 1) / 2 * 255)
                normal_img[20+i, 20+j] = [b, g, r]

    cv2.putText(normal_img, "Surface Normals (RGB = XYZ)", (10, 10), cv2.FONT_HERSHEY_SIMPLEX, 0.4, (255,255,255), 1)
    results.append(normal_img)
    titles.append("Surface Normals")
    print(f"    ✓ Surface normals visualized")

    # 3. Shape from Shading - Sphere
    print("\n[3] Shape from Shading - Sphere")
    sphere_img = np.zeros((200, 200), dtype=np.uint8)

    for i in range(200):
        for j in range(200):
            x = (i - 100) / 80
            y = (j - 100) / 80
            if x*x + y*y < 1:
                z = np.sqrt(max(0, 1 - x*x - y*y))
                lx, ly, lz = 0.5, -0.5, 0.7
                intensity = max(0, x*lx + y*ly + z*lz)
                sphere_img[i, j] = int(min(255, intensity * 255))

    cv2.putText(sphere_img, "Shaded Sphere", (10, 20), cv2.FONT_HERSHEY_SIMPLEX, 0.4, 255, 1)
    results.append(sphere_img)
    titles.append("Shaded Sphere")
    print(f"    ✓ Sphere with Lambertian shading created")

    # 4. Shape Recovery from Shading - FIXED
    print("\n[4] Shape Recovery from Shading")
    recovered = np.zeros((200, 200), dtype=np.uint8)

    h, w = sphere_img.shape
    for i in range(1, h-1):
        for j in range(1, w-1):
            dx = float(sphere_img[i+1, j]) - float(sphere_img[i-1, j])
            dy = float(sphere_img[i, j+1]) - float(sphere_img[i, j-1])
            if dx != 0 or dy != 0:
                val = np.sqrt(dx*dx + dy*dy) * 0.5
                recovered[i, j] = int(min(255, val))

    recovered = np.clip(recovered, 0, 255).astype(np.uint8)
    cv2.putText(recovered, "Recovered Shape", (10, 20), cv2.FONT_HERSHEY_SIMPLEX, 0.4, 255, 1)
    results.append(recovered)
    titles.append("Recovered Shape")
    print(f"    ✓ Shape recovered from shading")

    return results, titles

# ==================== 7. PHOTOMETRIC STEREO ====================

def demonstrate_photometric_stereo():
    """
    Demonstrate Photometric Stereo:
    1. Multiple Light Sources
    2. Normal Estimation
    3. Albedo Computation
    4. 3D Shape Recovery
    """
    print("\n" + "="*70)
    print("7. PHOTOMETRIC STEREO")
    print("="*70)

    print("\n[Mathematical Model]")
    print("Photometric Stereo:")
    print("[I1]   [L1x L1y L1z] [nx]")
    print("[I2] = [L2x L2y L2z] [ny]")
    print("[I3]   [L3x L3y L3z] [nz]")
    print("n = (L^T * L)^(-1) * L^T * I")

    results = []
    titles = []

    # 1. Multiple Light Sources - FIXED
    print("\n[1] Multiple Light Sources")
    light_img = np.zeros((350, 600, 3), dtype=np.uint8)
    light_img[:, :, :] = 20

    light_directions = [(-1, 0), (1, 0), (0, -1), (0, 1)]

    for light_idx, (lx, ly) in enumerate(light_directions):
        sphere = np.zeros((80, 80), dtype=np.uint8)
        for i in range(80):
            for j in range(80):
                x = (i - 40) / 30
                y = (j - 40) / 30
                if x*x + y*y < 1:
                    z = np.sqrt(max(0, 1 - x*x - y*y))
                    intensity = max(0, x*lx + y*ly + z*0.5)
                    sphere[i, j] = int(min(255, intensity * 255))

        row = light_idx // 2
        col = light_idx % 2
        x_offset = 50 + col * 250
        y_offset = 50 + row * 150

        # Convert grayscale to BGR
        sphere_color = cv2.cvtColor(sphere, cv2.COLOR_GRAY2BGR)
        light_img[y_offset:y_offset+80, x_offset:x_offset+80] = sphere_color

        end_x = x_offset + 40 + int(lx * 30)
        end_y = y_offset + 40 + int(ly * 30)
        cv2.arrowedLine(light_img, (x_offset+40, y_offset+40), (end_x, end_y), (0,255,255), 2)
        cv2.putText(light_img, f"Light {light_idx+1}", (x_offset+10, y_offset+10), cv2.FONT_HERSHEY_SIMPLEX, 0.4, (255,255,255), 1)

    cv2.putText(light_img, "Photometric Stereo: Multiple Light Sources", (10, 30), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (255,255,255), 1)
    results.append(light_img)
    titles.append("Multiple Light Sources")
    print(f"    ✓ Multiple light sources visualized")

    # 2. Surface Normal Estimation
    print("\n[2] Surface Normal Estimation")
    normal_est_img = np.zeros((300, 400, 3), dtype=np.uint8)
    normal_est_img[:, :, :] = 20

    for i in range(200):
        for j in range(200):
            x = (i - 100) / 80
            y = (j - 100) / 80
            if x*x + y*y < 0.9:
                z = np.sqrt(max(0, 1 - x*x - y*y))
                norm = max(0.01, np.sqrt(x*x + y*y + z*z))
                nx = x / norm
                ny = y / norm
                nz = z / norm

                r = int((nx + 1) / 2 * 255)
                g = int((ny + 1) / 2 * 255)
                b = int((nz + 1) / 2 * 255)
                normal_est_img[50+i, 50+j] = [b, g, r]

    cv2.putText(normal_est_img, "Estimated Surface Normals", (10, 30), cv2.FONT_HERSHEY_SIMPLEX, 0.4, (255,255,255), 1)
    results.append(normal_est_img)
    titles.append("Estimated Normals")
    print(f"    ✓ Surface normals estimated")

    # 3. Albedo Computation
    print("\n[3] Albedo Computation")
    albedo_img = np.zeros((200, 200), dtype=np.uint8)

    for i in range(200):
        for j in range(200):
            if i < 100 and j < 100:
                albedo_img[i, j] = 200
            elif i < 100 and j >= 100:
                albedo_img[i, j] = 150
            elif i >= 100 and j < 100:
                albedo_img[i, j] = 100
            else:
                albedo_img[i, j] = 50

    cv2.putText(albedo_img, "Albedo Map", (10, 20), cv2.FONT_HERSHEY_SIMPLEX, 0.4, 255, 1)
    results.append(albedo_img)
    titles.append("Albedo Map")
    print(f"    ✓ Albedo map computed")

    # 4. 3D Shape Recovery
    print("\n[4] 3D Shape Recovery from Photometric Stereo")
    shape3d_img = np.zeros((300, 400, 3), dtype=np.uint8)
    shape3d_img[:, :, :] = 20

    height_map = np.zeros((100, 100))
    for i in range(100):
        for j in range(100):
            x = (i - 50) / 40
            y = (j - 50) / 40
            if x*x + y*y < 1:
                height_map[i, j] = np.sqrt(max(0, 1 - x*x - y*y)) * 100

    for i in range(100):
        for j in range(100):
            h = int(height_map[i, j])
            if h > 0:
                color = (min(255, h*2), min(255, h), 50)
                x_pos = 150 + i * 2
                y_pos = 150 + j * 2 - h // 2
                if 0 <= x_pos < 400 and 0 <= y_pos < 300:
                    cv2.circle(shape3d_img, (x_pos, y_pos), 1, color, -1)

    cv2.putText(shape3d_img, "3D Shape from Photometric Stereo", (10, 30), cv2.FONT_HERSHEY_SIMPLEX, 0.4, (255,255,255), 1)
    results.append(shape3d_img)
    titles.append("3D Recovered Shape")
    print(f"    ✓ 3D shape recovered from photometric stereo")

    # 5. Photometric Stereo Pipeline
    print("\n[5] Photometric Stereo Pipeline")
    pipeline_img = np.zeros((350, 600, 3), dtype=np.uint8)
    pipeline_img[:, :, :] = 20

    pipeline_text = [
        "Photometric Stereo Pipeline:",
        "",
        "1. Capture images with different light directions",
        "2. Estimate surface normals using least squares",
        "3. Compute albedo from normals",
        "4. Integrate normals to get depth map",
        "5. Recover 3D shape",
        "",
        "Advantages:",
        "• High-quality 3D reconstruction",
        "• Works with textureless surfaces",
        "• Fast and efficient"
    ]

    for i, line in enumerate(pipeline_text):
        cv2.putText(pipeline_img, line, (20, 30 + i*30), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0,255,255), 1)

    results.append(pipeline_img)
    titles.append("Photometric Stereo Pipeline")
    print(f"    ✓ Photometric stereo pipeline explained")

    return results, titles

# ==================== COMPLETE DEMONSTRATION ====================

def display_results_grid(results, titles, title_prefix=""):
    """Display results in a grid"""
    if not results:
        return

    cols = min(4, len(results))
    rows = (len(results) + cols - 1) // cols

    fig, axes = plt.subplots(rows, cols, figsize=(20, rows * 4))
    if rows == 1:
        axes = axes.reshape(1, -1)

    for idx, (img_result, title) in enumerate(zip(results, titles)):
        row = idx // cols
        col = idx % cols
        if row < rows and col < cols:
            if len(img_result.shape) == 3:
                axes[row, col].imshow(img_result)
            else:
                axes[row, col].imshow(img_result, cmap='gray')
            axes[row, col].set_title(title, fontsize=10)
            axes[row, col].axis('off')

    for idx in range(len(results), rows * cols):
        row = idx // cols
        col = idx % cols
        if row < rows and col < cols:
            axes[row, col].axis('off')

    plt.suptitle(title_prefix, fontsize=14, y=1.02)
    plt.tight_layout()
    plt.show()

def create_test_image(rows=256, cols=256):
    """Create a synthetic test image"""
    img = np.zeros((rows, cols), dtype=np.uint8)

    grad = np.arange(rows).reshape(-1, 1) / rows * 255
    center_i, center_j = rows/2, cols/2
    y, x = np.ogrid[:rows, :cols]
    dist = np.sqrt((x - center_j)**2 + (y - center_i)**2)
    circle = 200 * (dist < 50)
    sin_pattern = 128 + 100 * np.sin(y/15) * np.cos(x/15)
    square = np.zeros((rows, cols))
    square[50:100, 50:100] = 150
    noise = np.random.randint(-20, 21, (rows, cols))

    img = ((grad + circle + sin_pattern + square + noise) / 4).astype(np.uint8)
    return np.clip(img, 0, 255).astype(np.uint8)

def main():
    print("\n" + "="*70)
    print("3D COMPUTER VISION DEMONSTRATION")
    print("Topics: AR/VR, Camera Calibration, Binocular Vision, Multiple View Geometry")
    print("="*70)
    print("\n✅ All operations use mathematical formulas from scratch")
    print("✅ Supports any image format (JPG, PNG, BMP, TIFF, GIF, WEBP, PGM)")
    print("✅ Auto-resizes to 720p max")
    print("\n" + "="*70)

    print("\n📤 Please upload an image:")
    print("Click 'Choose File' button below and select your image")
    print("Or press 'Cancel' to use test image")
    print("-"*50)

    uploaded = files.upload()

    if not uploaded:
        print("\n⚠ No file uploaded. Using test image...")
        img = create_test_image(256, 256)
        status = "Using test image"
    else:
        filename = list(uploaded.keys())[0]
        content = uploaded[filename]

        pil_img = Image.open(io.BytesIO(content))
        if pil_img.mode != 'L':
            pil_img = pil_img.convert('L')
        img = np.array(pil_img, dtype=np.uint8)
        status = f"✓ Loaded: {filename} ({img.shape[1]}x{img.shape[0]})"

    print(f"\n{status}")

    max_dim = 720
    h, w = img.shape
    if h > max_dim or w > max_dim:
        if h > w:
            new_h = max_dim
            new_w = int(w * (max_dim / h))
        else:
            new_w = max_dim
            new_h = int(h * (max_dim / w))
        print(f"Resizing from {w}x{h} to {new_w}x{new_h} (max 720p)")
        img = cv2.resize(img, (new_w, new_h))

    # ===== 1. AR vs VR =====
    print("\n" + "="*70)
    print("1. AUGMENTED REALITY VS VIRTUAL REALITY")
    print("="*70)
    results1, titles1 = demonstrate_ar_vs_vr()
    display_results_grid(results1, titles1, "AR vs VR")

    # ===== 2. CAMERA MODEL & CALIBRATION =====
    print("\n" + "="*70)
    print("2. CAMERA MODEL AND CALIBRATION")
    print("="*70)
    results2, titles2 = demonstrate_camera_model_calibration(img)
    display_results_grid(results2, titles2, "Camera Model & Calibration")

    # ===== 3. BINOCULAR IMAGING SYSTEM =====
    print("\n" + "="*70)
    print("3. BINOCULAR IMAGING SYSTEM")
    print("="*70)
    results3, titles3 = demonstrate_binocular_system(img)
    display_results_grid(results3, titles3, "Binocular Imaging System")

    # ===== 4. MULTIPLE VIEW GEOMETRY =====
    print("\n" + "="*70)
    print("4. MULTIPLE VIEW GEOMETRY")
    print("="*70)
    results4, titles4 = demonstrate_multiple_view_geometry()
    display_results_grid(results4, titles4, "Multiple View Geometry")

    # ===== 5. STRUCTURE DETERMINATION =====
    print("\n" + "="*70)
    print("5. STRUCTURE DETERMINATION")
    print("="*70)
    results5, titles5 = demonstrate_structure_determination()
    display_results_grid(results5, titles5, "Structure Determination")

    # ===== 6. SHAPE FROM SHADING =====
    print("\n" + "="*70)
    print("6. SHAPE FROM SHADING")
    print("="*70)
    results6, titles6 = demonstrate_shape_from_shading(img)
    display_results_grid(results6, titles6, "Shape from Shading")

    # ===== 7. PHOTOMETRIC STEREO =====
    print("\n" + "="*70)
    print("7. PHOTOMETRIC STEREO")
    print("="*70)
    results7, titles7 = demonstrate_photometric_stereo()
    display_results_grid(results7, titles7, "Photometric Stereo")

    # ===== SUMMARY =====
    print("\n" + "="*70)
    print("SUMMARY OF ALL DEMONSTRATIONS")
    print("="*70)

    print("\n1. Augmented Reality vs Virtual Reality:")
    print("   - AR: Real world + Virtual overlay")
    print("   - VR: Fully virtual environment")

    print("\n2. Camera Model and Calibration:")
    print("   - Pinhole Model: [x'] = [f 0 cx] [X]")
    print("   - Intrinsic Parameters: fx, fy, cx, cy")

    print("\n3. Binocular Imaging System:")
    print("   - Depth from Disparity: Z = (f*B)/d")
    print("   - Epipolar Geometry: x'^T * F * x = 0")

    print("\n4. Multiple View Geometry:")
    print("   - Fundamental Matrix F: Epipolar constraint")
    print("   - Essential Matrix E: Calibrated cameras")
    print("   - Homography H: Plane-to-plane mapping")

    print("\n5. Structure Determination:")
    print("   - Triangulation: 3D from 2D correspondences")
    print("   - 3D Reconstruction from multiple views")

    print("\n6. Shape from Shading:")
    print("   - Lambertian Model: I = ρ * (L · N)")
    print("   - Surface Normals from intensity")

    print("\n7. Photometric Stereo:")
    print("   - Multiple light sources")
    print("   - Normal estimation using least squares")
    print("   - 3D shape recovery")

    print("\n" + "="*70)
    save = input("Save all results to files? (y/n): ").strip().lower()
    if save == 'y':
        print("\nSaving images...")
        all_results = [
            (results1, titles1, "ar_vr"),
            (results2, titles2, "camera"),
            (results3, titles3, "binocular"),
            (results4, titles4, "geometry"),
            (results5, titles5, "structure"),
            (results6, titles6, "shading"),
            (results7, titles7, "photometric")
        ]

        for res_group, titles_group, prefix in all_results:
            for idx, (img_result, title) in enumerate(zip(res_group, titles_group)):
                safe_title = title.replace(' ', '_').replace('/', '_').replace('(', '').replace(')', '').replace('=', '_')
                filename = f"{prefix}_{idx:02d}_{safe_title}.png"
                if len(img_result.shape) == 3:
                    cv2.imwrite(filename, cv2.cvtColor(img_result, cv2.COLOR_RGB2BGR))
                else:
                    cv2.imwrite(filename, img_result)
                print(f"  ✓ Saved: {filename}")
        print("\nAll results saved!")

if __name__ == "__main__":
    main()
```
