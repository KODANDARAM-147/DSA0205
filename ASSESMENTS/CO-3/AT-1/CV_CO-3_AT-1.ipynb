```python
# ============================================
# AT1 - EXPERIMENT 1
# Image Preprocessing Pipeline:
# Noise Removal, Resizing, and Normalization
# ============================================

# Import required libraries
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# --------------------------------------------
# STEP 1: Upload Image
# --------------------------------------------

print("Upload an image:")
uploaded = files.upload()

# Get uploaded file name
filename = next(iter(uploaded))

# Read image
image = cv2.imread(filename)

# Check whether image was loaded successfully
if image is None:
    print("Error: Unable to read the uploaded image.")
else:

    # Convert BGR to RGB for displaying with Matplotlib
    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    # --------------------------------------------
    # STEP 2: Add Gaussian Noise
    # --------------------------------------------

    # Generate Gaussian noise
    mean = 0
    sigma = 25

    noise = np.random.normal(
        mean,
        sigma,
        image_rgb.shape
    ).astype(np.float32)

    # Add noise to image
    noisy_image = image_rgb.astype(np.float32) + noise

    # Keep pixel values between 0 and 255
    noisy_image = np.clip(noisy_image, 0, 255).astype(np.uint8)

    # --------------------------------------------
    # STEP 3: Noise Removal
    # --------------------------------------------

    # Apply Gaussian Blur
    denoised_image = cv2.GaussianBlur(
        noisy_image,
        (5, 5),
        0
    )

    # --------------------------------------------
    # STEP 4: Image Resizing
    # --------------------------------------------

    resized_image = cv2.resize(
        denoised_image,
        (256, 256)
    )

    # --------------------------------------------
    # STEP 5: Normalization
    # --------------------------------------------

    # Convert pixel values from [0,255] to [0,1]
    normalized_image = resized_image.astype(
        np.float32
    ) / 255.0

    # --------------------------------------------
    # STEP 6: Display Results
    # --------------------------------------------

    plt.figure(figsize=(16, 10))

    # Original Image
    plt.subplot(2, 3, 1)
    plt.imshow(image_rgb)
    plt.title("Original Image")
    plt.axis("off")

    # Noisy Image
    plt.subplot(2, 3, 2)
    plt.imshow(noisy_image)
    plt.title("Noisy Image")
    plt.axis("off")

    # Denoised Image
    plt.subplot(2, 3, 3)
    plt.imshow(denoised_image)
    plt.title("Denoised Image")
    plt.axis("off")

    # Resized Image
    plt.subplot(2, 3, 4)
    plt.imshow(resized_image)
    plt.title("Resized Image (256 × 256)")
    plt.axis("off")

    # Normalized Image
    plt.subplot(2, 3, 5)
    plt.imshow(normalized_image)
    plt.title("Normalized Image")
    plt.axis("off")

    plt.tight_layout()
    plt.show()

    # --------------------------------------------
    # STEP 7: Display Image Information
    # --------------------------------------------

    print("========== IMAGE INFORMATION ==========")

    print("Original image shape:", image_rgb.shape)
    print("Noisy image shape:", noisy_image.shape)
    print("Denoised image shape:", denoised_image.shape)
    print("Resized image shape:", resized_image.shape)

    print("\nOriginal pixel range:",
          image_rgb.min(), "to", image_rgb.max())

    print("Normalized pixel range:",
          normalized_image.min(), "to",
          normalized_image.max())

    print("\nNormalization completed successfully.")
```

```python
# ==========================================================
# AT1 - EXPERIMENT 3
# Edge Detection Using the Sobel Operator
# ==========================================================

# Import required libraries
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# ----------------------------------------------------------
# STEP 1: Upload Image
# ----------------------------------------------------------

print("Upload an image:")
uploaded = files.upload()

# Get uploaded file name
filename = next(iter(uploaded))

# Read the image
image = cv2.imread(filename)

# Check whether image was loaded
if image is None:
    print("Error: Unable to read the uploaded image.")

else:

    # ------------------------------------------------------
    # STEP 2: Convert Image to RGB
    # ------------------------------------------------------

    rgb_image = cv2.cvtColor(
        image,
        cv2.COLOR_BGR2RGB
    )

    # ------------------------------------------------------
    # STEP 3: Convert RGB Image to Grayscale
    # ------------------------------------------------------

    gray = cv2.cvtColor(
        rgb_image,
        cv2.COLOR_RGB2GRAY
    )

    # ------------------------------------------------------
    # STEP 4: Apply Sobel Operator in X Direction
    # ------------------------------------------------------

    sobel_x = cv2.Sobel(
        gray,
        cv2.CV_64F,
        1,
        0,
        ksize=3
    )

    # Convert to absolute scale
    sobel_x = cv2.convertScaleAbs(sobel_x)

    # ------------------------------------------------------
    # STEP 5: Apply Sobel Operator in Y Direction
    # ------------------------------------------------------

    sobel_y = cv2.Sobel(
        gray,
        cv2.CV_64F,
        0,
        1,
        ksize=3
    )

    # Convert to absolute scale
    sobel_y = cv2.convertScaleAbs(sobel_y)

    # ------------------------------------------------------
    # STEP 6: Calculate Gradient Magnitude
    # ------------------------------------------------------

    # Calculate gradient using X and Y components
    sobel_combined = cv2.magnitude(
        cv2.Sobel(gray, cv2.CV_64F, 1, 0, ksize=3),
        cv2.Sobel(gray, cv2.CV_64F, 0, 1, ksize=3)
    )

    # Normalize gradient magnitude for display
    sobel_combined = cv2.normalize(
        sobel_combined,
        None,
        0,
        255,
        cv2.NORM_MINMAX
    ).astype(np.uint8)

    # ------------------------------------------------------
    # STEP 7: Display Results
    # ------------------------------------------------------

    plt.figure(figsize=(15, 10))

    # Original image
    plt.subplot(2, 3, 1)
    plt.imshow(rgb_image)
    plt.title("Original Image")
    plt.axis("off")

    # Grayscale image
    plt.subplot(2, 3, 2)
    plt.imshow(gray, cmap="gray")
    plt.title("Grayscale Image")
    plt.axis("off")

    # Sobel X
    plt.subplot(2, 3, 3)
    plt.imshow(sobel_x, cmap="gray")
    plt.title("Sobel X - Vertical Edges")
    plt.axis("off")

    # Sobel Y
    plt.subplot(2, 3, 4)
    plt.imshow(sobel_y, cmap="gray")
    plt.title("Sobel Y - Horizontal Edges")
    plt.axis("off")

    # Combined Sobel
    plt.subplot(2, 3, 5)
    plt.imshow(sobel_combined, cmap="gray")
    plt.title("Combined Sobel Edges")
    plt.axis("off")

    plt.tight_layout()
    plt.show()

    # ------------------------------------------------------
    # STEP 8: Display Information
    # ------------------------------------------------------

    print("========== SOBEL EDGE DETECTION ==========")

    print("Original image shape:", rgb_image.shape)
    print("Grayscale image shape:", gray.shape)

    print("\nSobel X:")
    print("Detects intensity changes mainly in the X direction.")

    print("\nSobel Y:")
    print("Detects intensity changes mainly in the Y direction.")

    print("\nCombined Sobel:")
    print("Combines gradients from both X and Y directions.")

    print("\nSobel edge detection completed successfully.")
```

```python
# ==========================================================
# AT1 - EXPERIMENT 5
# Corner Detection Using the Harris Corner Detector
# ==========================================================

# Import required libraries
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# ----------------------------------------------------------
# STEP 1: Upload Image
# ----------------------------------------------------------

print("Upload an image:")
uploaded = files.upload()

# Get uploaded filename
filename = next(iter(uploaded))

# Read image
image = cv2.imread(filename)

# Check image
if image is None:
    print("Error: Unable to read the uploaded image.")

else:

    # ------------------------------------------------------
    # STEP 2: Convert BGR to RGB
    # ------------------------------------------------------

    rgb_image = cv2.cvtColor(
        image,
        cv2.COLOR_BGR2RGB
    )

    # ------------------------------------------------------
    # STEP 3: Convert Image to Grayscale
    # ------------------------------------------------------

    gray = cv2.cvtColor(
        rgb_image,
        cv2.COLOR_RGB2GRAY
    )

    # ------------------------------------------------------
    # STEP 4: Convert to Float32
    # Harris requires floating-point image
    # ------------------------------------------------------

    gray_float = np.float32(gray)

    # ------------------------------------------------------
    # STEP 5: Apply Harris Corner Detector
    # ------------------------------------------------------

    harris_response = cv2.cornerHarris(
        gray_float,
        blockSize=2,
        ksize=3,
        k=0.04
    )

    # ------------------------------------------------------
    # STEP 6: Dilate Harris Response
    # This makes corner points easier to visualize
    # ------------------------------------------------------

    harris_response = cv2.dilate(
        harris_response,
        None
    )

    # ------------------------------------------------------
    # STEP 7: Threshold the Harris Response
    # ------------------------------------------------------

    threshold = 0.01 * harris_response.max()

    corner_mask = (
        harris_response > threshold
    )

    # ------------------------------------------------------
    # STEP 8: Mark Detected Corners
    # ------------------------------------------------------

    result = rgb_image.copy()

    result[corner_mask] = [255, 0, 0]

    # ------------------------------------------------------
    # STEP 9: Count Corners
    # ------------------------------------------------------

    corner_count = np.count_nonzero(
        corner_mask
    )

    # ------------------------------------------------------
    # STEP 10: Display Results
    # ------------------------------------------------------

    plt.figure(figsize=(15, 10))

    # Original image
    plt.subplot(2, 2, 1)
    plt.imshow(rgb_image)
    plt.title("Original Image")
    plt.axis("off")

    # Grayscale image
    plt.subplot(2, 2, 2)
    plt.imshow(gray, cmap="gray")
    plt.title("Grayscale Image")
    plt.axis("off")

    # Harris response
    plt.subplot(2, 2, 3)
    plt.imshow(
        harris_response,
        cmap="gray"
    )
    plt.title("Harris Corner Response")
    plt.axis("off")

    # Detected corners
    plt.subplot(2, 2, 4)
    plt.imshow(result)
    plt.title("Detected Harris Corners")
    plt.axis("off")

    plt.tight_layout()
    plt.show()

    # ------------------------------------------------------
    # STEP 11: Display Information
    # ------------------------------------------------------

    print("========== HARRIS CORNER DETECTION ==========")

    print("Original image shape:", rgb_image.shape)

    print("Grayscale image shape:", gray.shape)

    print("Harris block size:", 2)

    print("Harris kernel size:", 3)

    print("Harris k value:", 0.04)

    print("Corner threshold:",
          threshold)

    print("Detected corner pixels:",
          corner_count)

    print("\nHarris corner detection completed successfully.")
```

```python
# ==========================================================
# AT1 - EXPERIMENT 6
# Corner Response Analysis Across Varying Threshold Values
# ==========================================================

# Import required libraries
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# ----------------------------------------------------------
# STEP 1: Upload Image
# ----------------------------------------------------------

print("Upload an image:")
uploaded = files.upload()

# Get uploaded filename
filename = next(iter(uploaded))

# Read image
image = cv2.imread(filename)

# Check whether image was loaded
if image is None:
    print("Error: Unable to read the uploaded image.")

else:

    # ------------------------------------------------------
    # STEP 2: Convert BGR to RGB
    # ------------------------------------------------------

    rgb_image = cv2.cvtColor(
        image,
        cv2.COLOR_BGR2RGB
    )

    # ------------------------------------------------------
    # STEP 3: Convert Image to Grayscale
    # ------------------------------------------------------

    gray = cv2.cvtColor(
        rgb_image,
        cv2.COLOR_RGB2GRAY
    )

    # Convert to float32
    gray_float = np.float32(gray)

    # ------------------------------------------------------
    # STEP 4: Apply Harris Corner Detector
    # ------------------------------------------------------

    harris_response = cv2.cornerHarris(
        gray_float,
        blockSize=2,
        ksize=3,
        k=0.04
    )

    # Dilate response for better visualization
    harris_response = cv2.dilate(
        harris_response,
        None
    )

    # ------------------------------------------------------
    # STEP 5: Find Maximum Harris Response
    # ------------------------------------------------------

    max_response = harris_response.max()

    # ------------------------------------------------------
    # STEP 6: Define Different Threshold Values
    # ------------------------------------------------------

    threshold_percentages = [
        0.005,
        0.01,
        0.02,
        0.05
    ]

    # Store results
    corner_counts = []
    corner_images = []

    # ------------------------------------------------------
    # STEP 7: Apply Different Thresholds
    # ------------------------------------------------------

    for threshold_percentage in threshold_percentages:

        # Calculate threshold
        threshold = (
            threshold_percentage *
            max_response
        )

        # Create corner mask
        corner_mask = (
            harris_response > threshold
        )

        # Count detected corner pixels
        count = np.count_nonzero(
            corner_mask
        )

        corner_counts.append(count)

        # Copy original image
        result = rgb_image.copy()

        # Mark detected corners in red
        result[corner_mask] = [255, 0, 0]

        # Store image
        corner_images.append(result)

    # ------------------------------------------------------
    # STEP 8: Display Results
    # ------------------------------------------------------

    plt.figure(figsize=(16, 12))

    # Original image
    plt.subplot(2, 3, 1)
    plt.imshow(rgb_image)
    plt.title("Original Image")
    plt.axis("off")

    # Threshold 0.5%
    plt.subplot(2, 3, 2)
    plt.imshow(corner_images[0])
    plt.title("Threshold = 0.5%")
    plt.axis("off")

    # Threshold 1%
    plt.subplot(2, 3, 3)
    plt.imshow(corner_images[1])
    plt.title("Threshold = 1%")
    plt.axis("off")

    # Threshold 2%
    plt.subplot(2, 3, 4)
    plt.imshow(corner_images[2])
    plt.title("Threshold = 2%")
    plt.axis("off")

    # Threshold 5%
    plt.subplot(2, 3, 5)
    plt.imshow(corner_images[3])
    plt.title("Threshold = 5%")
    plt.axis("off")

    plt.tight_layout()
    plt.show()

    # ------------------------------------------------------
    # STEP 9: Display Harris Response
    # ------------------------------------------------------

    plt.figure(figsize=(7, 5))

    plt.imshow(
        harris_response,
        cmap="gray"
    )

    plt.title("Harris Corner Response")
    plt.colorbar()
    plt.axis("off")
    plt.show()

    # ------------------------------------------------------
    # STEP 10: Display Numerical Results
    # ------------------------------------------------------

    print("========== HARRIS THRESHOLD ANALYSIS ==========")

    print("Image shape:", rgb_image.shape)

    print("Harris block size:", 2)
    print("Harris kernel size:", 3)
    print("Harris k value:", 0.04)

    print("\nThreshold Analysis:")

    for i in range(len(threshold_percentages)):

        percentage = (
            threshold_percentages[i] * 100
        )

        print(
            f"Threshold = {percentage:.1f}% "
            f"--> Detected corner pixels = "
            f"{corner_counts[i]}"
        )

    # ------------------------------------------------------
    # STEP 11: Plot Threshold vs Corner Count
    # ------------------------------------------------------

    threshold_labels = [
        percentage * 100
        for percentage in threshold_percentages
    ]

    plt.figure(figsize=(8, 5))

    plt.plot(
        threshold_labels,
        corner_counts,
        marker="o"
    )

    plt.xlabel("Threshold (%)")
    plt.ylabel("Detected Corner Pixels")
    plt.title(
        "Harris Corner Response Analysis"
    )

    plt.grid(True)
    plt.show()

    print("\nThreshold analysis completed successfully.")
```

```python
# ==========================================================
# AT1 - EXPERIMENT 7
# Line Detection Using the Hough Transform
# ==========================================================

# Import required libraries
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# ----------------------------------------------------------
# STEP 1: Upload Image
# ----------------------------------------------------------

print("Upload an image:")
uploaded = files.upload()

# Get uploaded filename
filename = next(iter(uploaded))

# Read image
image = cv2.imread(filename)

# Check image
if image is None:
    print("Error: Unable to read the uploaded image.")

else:

    # ------------------------------------------------------
    # STEP 2: Convert BGR to RGB
    # ------------------------------------------------------

    rgb_image = cv2.cvtColor(
        image,
        cv2.COLOR_BGR2RGB
    )

    # ------------------------------------------------------
    # STEP 3: Convert Image to Grayscale
    # ------------------------------------------------------

    gray = cv2.cvtColor(
        rgb_image,
        cv2.COLOR_RGB2GRAY
    )

    # ------------------------------------------------------
    # STEP 4: Apply Gaussian Blur
    # ------------------------------------------------------

    blurred = cv2.GaussianBlur(
        gray,
        (5, 5),
        0
    )

    # ------------------------------------------------------
    # STEP 5: Detect Edges Using Canny
    # ------------------------------------------------------

    edges = cv2.Canny(
        blurred,
        50,
        150
    )

    # ------------------------------------------------------
    # STEP 6: Apply Standard Hough Line Transform
    # ------------------------------------------------------

    lines = cv2.HoughLines(
        edges,
        1,
        np.pi / 180,
        100
    )

    # Create a copy for drawing lines
    result = rgb_image.copy()

    # Count detected lines
    line_count = 0

    # ------------------------------------------------------
    # STEP 7: Draw Detected Lines
    # ------------------------------------------------------

    if lines is not None:

        for line in lines:

            rho, theta = line[0]

            # Calculate cosine and sine
            a = np.cos(theta)
            b = np.sin(theta)

            # Point on the detected line
            x0 = a * rho
            y0 = b * rho

            # Large value used to extend line
            scale = 1000

            # Calculate two points
            x1 = int(x0 + scale * (-b))
            y1 = int(y0 + scale * (a))

            x2 = int(x0 - scale * (-b))
            y2 = int(y0 - scale * (a))

            # Draw line
            cv2.line(
                result,
                (x1, y1),
                (x2, y2),
                (255, 0, 0),
                2
            )

            line_count += 1

    # ------------------------------------------------------
    # STEP 8: Display Results
    # ------------------------------------------------------

    plt.figure(figsize=(16, 10))

    # Original image
    plt.subplot(2, 2, 1)
    plt.imshow(rgb_image)
    plt.title("Original Image")
    plt.axis("off")

    # Grayscale image
    plt.subplot(2, 2, 2)
    plt.imshow(gray, cmap="gray")
    plt.title("Grayscale Image")
    plt.axis("off")

    # Edge image
    plt.subplot(2, 2, 3)
    plt.imshow(edges, cmap="gray")
    plt.title("Canny Edge Image")
    plt.axis("off")

    # Hough lines
    plt.subplot(2, 2, 4)
    plt.imshow(result)
    plt.title("Detected Hough Lines")
    plt.axis("off")

    plt.tight_layout()
    plt.show()

    # ------------------------------------------------------
    # STEP 9: Display Information
    # ------------------------------------------------------

    print("========== HOUGH LINE DETECTION ==========")

    print("Original image shape:", rgb_image.shape)

    print("Edge image shape:", edges.shape)

    if lines is not None:
        print("Number of detected Hough lines:",
              line_count)
    else:
        print("No Hough lines detected.")

    print("\nHough line detection completed successfully.")
```

```python
# ==========================================================
# AT1 - EXPERIMENT 8
# Circle Detection Using the Hough Circle Transform
# ==========================================================

# Import required libraries
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# ----------------------------------------------------------
# STEP 1: Upload Image
# ----------------------------------------------------------

print("Upload an image:")
uploaded = files.upload()

# Get uploaded filename
filename = next(iter(uploaded))

# Read image
image = cv2.imread(filename)

# Check image
if image is None:
    print("Error: Unable to read the uploaded image.")

else:

    # ------------------------------------------------------
    # STEP 2: Convert BGR to RGB
    # ------------------------------------------------------

    rgb_image = cv2.cvtColor(
        image,
        cv2.COLOR_BGR2RGB
    )

    # ------------------------------------------------------
    # STEP 3: Convert Image to Grayscale
    # ------------------------------------------------------

    gray = cv2.cvtColor(
        rgb_image,
        cv2.COLOR_RGB2GRAY
    )

    # ------------------------------------------------------
    # STEP 4: Apply Gaussian Blur
    # ------------------------------------------------------

    blurred = cv2.GaussianBlur(
        gray,
        (9, 9),
        2
    )

    # ------------------------------------------------------
    # STEP 5: Apply Hough Circle Transform
    # ------------------------------------------------------

    circles = cv2.HoughCircles(
        blurred,
        cv2.HOUGH_GRADIENT,
        dp=1.2,
        minDist=30,
        param1=100,
        param2=30,
        minRadius=10,
        maxRadius=200
    )

    # Create a copy for drawing
    result = rgb_image.copy()

    # ------------------------------------------------------
    # STEP 6: Draw Detected Circles
    # ------------------------------------------------------

    circle_count = 0

    if circles is not None:

        # Round the circle values
        circles = np.round(
            circles[0, :]
        ).astype(int)

        circle_count = len(circles)

        for x, y, r in circles:

            # Draw outer circle
            cv2.circle(
                result,
                (x, y),
                r,
                (255, 0, 0),
                3
            )

            # Draw center point
            cv2.circle(
                result,
                (x, y),
                3,
                (0, 255, 0),
                -1
            )

    # ------------------------------------------------------
    # STEP 7: Display Results
    # ------------------------------------------------------

    plt.figure(figsize=(16, 10))

    # Original image
    plt.subplot(2, 2, 1)
    plt.imshow(rgb_image)
    plt.title("Original Image")
    plt.axis("off")

    # Grayscale image
    plt.subplot(2, 2, 2)
    plt.imshow(gray, cmap="gray")
    plt.title("Grayscale Image")
    plt.axis("off")

    # Blurred image
    plt.subplot(2, 2, 3)
    plt.imshow(blurred, cmap="gray")
    plt.title("Gaussian Blurred Image")
    plt.axis("off")

    # Detected circles
    plt.subplot(2, 2, 4)
    plt.imshow(result)
    plt.title("Detected Hough Circles")
    plt.axis("off")

    plt.tight_layout()
    plt.show()

    # ------------------------------------------------------
    # STEP 8: Display Circle Information
    # ------------------------------------------------------

    print("========== HOUGH CIRCLE DETECTION ==========")

    print("Original image shape:",
          rgb_image.shape)

    print("Number of detected circles:",
          circle_count)

    if circles is not None:

        print("\nDetected Circle Parameters:")
        print("(Center X, Center Y, Radius)")

        for i, (x, y, r) in enumerate(
            circles,
            start=1
        ):
            print(
                f"Circle {i}: "
                f"Center=({x}, {y}), "
                f"Radius={r}"
            )

    else:
        print("\nNo circles were detected.")

    print("\nHough Circle Transform completed successfully.")
```

```python
# ==========================================================
# AT1 - EXPERIMENT 10
# Feature Matching Between Two Images Using SIFT
# and a Distance-Ratio Test
# ==========================================================

# Import required libraries
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# ----------------------------------------------------------
# STEP 1: Upload TWO Images
# ----------------------------------------------------------

print("Upload TWO related images:")
uploaded = files.upload()

# Check number of uploaded images
if len(uploaded) < 2:
    print("Error: Please upload exactly two images.")

else:

    # Get the first two uploaded filenames
    filenames = list(uploaded.keys())

    file1 = filenames[0]
    file2 = filenames[1]

    # ------------------------------------------------------
    # STEP 2: Read Both Images
    # ------------------------------------------------------

    image1 = cv2.imread(file1)
    image2 = cv2.imread(file2)

    # Check images
    if image1 is None or image2 is None:

        print("Error: Unable to read one or both images.")

    else:

        # --------------------------------------------------
        # STEP 3: Convert Images to RGB
        # --------------------------------------------------

        rgb1 = cv2.cvtColor(
            image1,
            cv2.COLOR_BGR2RGB
        )

        rgb2 = cv2.cvtColor(
            image2,
            cv2.COLOR_BGR2RGB
        )

        # --------------------------------------------------
        # STEP 4: Convert Images to Grayscale
        # --------------------------------------------------

        gray1 = cv2.cvtColor(
            rgb1,
            cv2.COLOR_RGB2GRAY
        )

        gray2 = cv2.cvtColor(
            rgb2,
            cv2.COLOR_RGB2GRAY
        )

        # --------------------------------------------------
        # STEP 5: Create SIFT Detector
        # --------------------------------------------------

        sift = cv2.SIFT_create()

        # --------------------------------------------------
        # STEP 6: Detect Keypoints and Descriptors
        # --------------------------------------------------

        keypoints1, descriptors1 = (
            sift.detectAndCompute(gray1, None)
        )

        keypoints2, descriptors2 = (
            sift.detectAndCompute(gray2, None)
        )

        # Check descriptors
        if descriptors1 is None or descriptors2 is None:

            print("Error: SIFT descriptors could not be generated.")

        else:

            # --------------------------------------------------
            # STEP 7: Create BFMatcher
            # --------------------------------------------------

            bf = cv2.BFMatcher(
                cv2.NORM_L2,
                crossCheck=False
            )

            # --------------------------------------------------
            # STEP 8: KNN Matching
            # Find two nearest matches
            # --------------------------------------------------

            knn_matches = bf.knnMatch(
                descriptors1,
                descriptors2,
                k=2
            )

            # --------------------------------------------------
            # STEP 9: Apply Lowe's Ratio Test
            # --------------------------------------------------

            ratio_threshold = 0.75

            good_matches = []

            for match_pair in knn_matches:

                # Make sure two matches exist
                if len(match_pair) == 2:

                    m, n = match_pair

                    # Distance-ratio test
                    if m.distance < ratio_threshold * n.distance:
                        good_matches.append(m)

            # --------------------------------------------------
            # STEP 10: Draw Good Matches
            # --------------------------------------------------

            matched_image = cv2.drawMatches(
                rgb1,
                keypoints1,
                rgb2,
                keypoints2,
                good_matches,
                None,
                flags=cv2.DrawMatchesFlags_NOT_DRAW_SINGLE_POINTS
            )

            # --------------------------------------------------
            # STEP 11: Display Input Images
            # --------------------------------------------------

            plt.figure(figsize=(15, 6))

            plt.subplot(1, 2, 1)
            plt.imshow(rgb1)
            plt.title("Image 1")
            plt.axis("off")

            plt.subplot(1, 2, 2)
            plt.imshow(rgb2)
            plt.title("Image 2")
            plt.axis("off")

            plt.tight_layout()
            plt.show()

            # --------------------------------------------------
            # STEP 12: Display Matching Result
            # --------------------------------------------------

            plt.figure(figsize=(18, 8))

            plt.imshow(matched_image)

            plt.title(
                "SIFT Feature Matching "
                "(Lowe's Ratio Test)"
            )

            plt.axis("off")
            plt.show()

            # --------------------------------------------------
            # STEP 13: Display Results
            # --------------------------------------------------

            print("========== SIFT FEATURE MATCHING ==========")

            print("\nImage 1:")
            print("Number of keypoints:",
                  len(keypoints1))

            print("\nImage 2:")
            print("Number of keypoints:",
                  len(keypoints2))

            print("\nTotal KNN match pairs:",
                  len(knn_matches))

            print("\nRatio threshold:",
                  ratio_threshold)

            print("\nGood matches after ratio test:",
                  len(good_matches))

            print(
                "\nSIFT feature matching completed successfully."
            )
```

```python
# ==========================================================
# AT1 - EXPERIMENT 11
# Dimensionality Reduction of Image Data Using PCA
# ==========================================================

# Import required libraries
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files
from sklearn.decomposition import PCA

# ----------------------------------------------------------
# STEP 1: Upload Image
# ----------------------------------------------------------

print("Upload an image:")
uploaded = files.upload()

# Get uploaded filename
filename = next(iter(uploaded))

# Read image
image = cv2.imread(filename)

# Check image
if image is None:
    print("Error: Unable to read the uploaded image.")

else:

    # ------------------------------------------------------
    # STEP 2: Convert Image to Grayscale
    # ------------------------------------------------------

    gray = cv2.cvtColor(
        image,
        cv2.COLOR_BGR2GRAY
    )

    # ------------------------------------------------------
    # STEP 3: Resize Image
    # ------------------------------------------------------

    resized = cv2.resize(
        gray,
        (128, 128)
    )

    # Convert pixel values to float
    image_float = resized.astype(
        np.float64
    ) / 255.0

    # ------------------------------------------------------
    # STEP 4: Apply PCA
    # ------------------------------------------------------
    # PCA is applied row-wise to the image matrix.
    # Each row is treated as one data sample.

    pca_full = PCA()

    transformed_full = pca_full.fit_transform(
        image_float
    )

    # ------------------------------------------------------
    # STEP 5: Calculate Cumulative Explained Variance
    # ------------------------------------------------------

    cumulative_variance = np.cumsum(
        pca_full.explained_variance_ratio_
    )

    # Find number of components required
    # to retain at least 95% variance

    components_95 = np.argmax(
        cumulative_variance >= 0.95
    ) + 1

    # ------------------------------------------------------
    # STEP 6: Apply PCA Using Selected Components
    # ------------------------------------------------------

    pca_reduced = PCA(
        n_components=components_95
    )

    transformed = pca_reduced.fit_transform(
        image_float
    )

    # ------------------------------------------------------
    # STEP 7: Reconstruct Image
    # ------------------------------------------------------

    reconstructed = pca_reduced.inverse_transform(
        transformed
    )

    # Keep pixel values within valid range
    reconstructed = np.clip(
        reconstructed,
        0,
        1
    )

    # ------------------------------------------------------
    # STEP 8: Calculate Variance Retained
    # ------------------------------------------------------

    retained_variance = (
        np.sum(
            pca_reduced.explained_variance_ratio_
        ) * 100
    )

    # ------------------------------------------------------
    # STEP 9: Calculate Compression Ratio
    # ------------------------------------------------------

    original_values = (
        image_float.shape[0] *
        image_float.shape[1]
    )

    reduced_values = (
        image_float.shape[0] *
        components_95
    )

    compression_ratio = (
        original_values /
        reduced_values
    )

    # ------------------------------------------------------
    # STEP 10: Display Original and Reconstructed Image
    # ------------------------------------------------------

    plt.figure(figsize=(15, 5))

    # Original
    plt.subplot(1, 3, 1)

    plt.imshow(
        image_float,
        cmap="gray"
    )

    plt.title(
        "Original Grayscale Image"
    )

    plt.axis("off")

    # Reconstructed
    plt.subplot(1, 3, 2)

    plt.imshow(
        reconstructed,
        cmap="gray"
    )

    plt.title(
        f"PCA Reconstructed Image\n"
        f"{components_95} Components"
    )

    plt.axis("off")

    # Difference
    difference = np.abs(
        image_float -
        reconstructed
    )

    plt.subplot(1, 3, 3)

    plt.imshow(
        difference,
        cmap="gray"
    )

    plt.title(
        "Reconstruction Difference"
    )

    plt.axis("off")

    plt.tight_layout()
    plt.show()

    # ------------------------------------------------------
    # STEP 11: Plot Explained Variance
    # ------------------------------------------------------

    plt.figure(figsize=(9, 5))

    plt.plot(
        range(
            1,
            len(cumulative_variance) + 1
        ),
        cumulative_variance * 100
    )

    plt.axhline(
        y=95,
        linestyle="--",
        label="95% Variance"
    )

    plt.axvline(
        x=components_95,
        linestyle="--",
        label=f"{components_95} Components"
    )

    plt.xlabel(
        "Number of Principal Components"
    )

    plt.ylabel(
        "Cumulative Explained Variance (%)"
    )

    plt.title(
        "PCA Explained Variance Analysis"
    )

    plt.legend()
    plt.grid(True)
    plt.show()

    # ------------------------------------------------------
    # STEP 12: Display PCA Results
    # ------------------------------------------------------

    print("========== PCA IMAGE ANALYSIS ==========")

    print(
        "Original image size:",
        image_float.shape
    )

    print(
        "Original dimensionality:",
        original_values
    )

    print(
        "Components required for 95% variance:",
        components_95
    )

    print(
        "Reduced representation size:",
        reduced_values
    )

    print(
        "Variance retained:",
        f"{retained_variance:.2f}%"
    )

    print(
        "Compression ratio:",
        f"{compression_ratio:.2f}:1"
    )

    print(
        "\nPCA dimensionality reduction completed successfully."
    )
```

```python
# ==========================================================
# AT1 - EXPERIMENT 12
# End-to-End Feature Pipeline:
# From Preprocessing to Matched Keypoints
# ==========================================================

# Import required libraries
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# ----------------------------------------------------------
# STEP 1: Upload TWO Images
# ----------------------------------------------------------

print("Upload TWO related images:")
uploaded = files.upload()

# Check number of images
if len(uploaded) < 2:

    print("Error: Please upload two images.")

else:

    # Get uploaded filenames
    filenames = list(uploaded.keys())

    file1 = filenames[0]
    file2 = filenames[1]

    # ------------------------------------------------------
    # STEP 2: Read Both Images
    # ------------------------------------------------------

    image1 = cv2.imread(file1)
    image2 = cv2.imread(file2)

    if image1 is None or image2 is None:

        print("Error: Unable to read one or both images.")

    else:

        # --------------------------------------------------
        # STEP 3: Convert BGR to RGB
        # --------------------------------------------------

        rgb1 = cv2.cvtColor(
            image1,
            cv2.COLOR_BGR2RGB
        )

        rgb2 = cv2.cvtColor(
            image2,
            cv2.COLOR_BGR2RGB
        )

        # --------------------------------------------------
        # STEP 4: Resize Images
        # --------------------------------------------------

        target_size = (640, 480)

        resized1 = cv2.resize(
            rgb1,
            target_size
        )

        resized2 = cv2.resize(
            rgb2,
            target_size
        )

        # --------------------------------------------------
        # STEP 5: Convert to Grayscale
        # --------------------------------------------------

        gray1 = cv2.cvtColor(
            resized1,
            cv2.COLOR_RGB2GRAY
        )

        gray2 = cv2.cvtColor(
            resized2,
            cv2.COLOR_RGB2GRAY
        )

        # --------------------------------------------------
        # STEP 6: Gaussian Noise Removal
        # --------------------------------------------------

        denoised1 = cv2.GaussianBlur(
            gray1,
            (5, 5),
            0
        )

        denoised2 = cv2.GaussianBlur(
            gray2,
            (5, 5),
            0
        )

        # --------------------------------------------------
        # STEP 7: Normalize Images
        # --------------------------------------------------

        normalized1 = cv2.normalize(
            denoised1,
            None,
            0,
            255,
            cv2.NORM_MINMAX
        )

        normalized2 = cv2.normalize(
            denoised2,
            None,
            0,
            255,
            cv2.NORM_MINMAX
        )

        # --------------------------------------------------
        # STEP 8: Create SIFT Detector
        # --------------------------------------------------

        sift = cv2.SIFT_create()

        # --------------------------------------------------
        # STEP 9: Detect Keypoints and Descriptors
        # --------------------------------------------------

        keypoints1, descriptors1 = (
            sift.detectAndCompute(
                normalized1,
                None
            )
        )

        keypoints2, descriptors2 = (
            sift.detectAndCompute(
                normalized2,
                None
            )
        )

        # Check descriptors
        if descriptors1 is None or descriptors2 is None:

            print(
                "Error: SIFT descriptors could not "
                "be generated."
            )

        else:

            # --------------------------------------------------
            # STEP 10: Create Brute-Force Matcher
            # --------------------------------------------------

            bf = cv2.BFMatcher(
                cv2.NORM_L2,
                crossCheck=False
            )

            # --------------------------------------------------
            # STEP 11: KNN Feature Matching
            # --------------------------------------------------

            knn_matches = bf.knnMatch(
                descriptors1,
                descriptors2,
                k=2
            )

            # --------------------------------------------------
            # STEP 12: Lowe's Ratio Test
            # --------------------------------------------------

            ratio_threshold = 0.75

            good_matches = []

            for pair in knn_matches:

                if len(pair) == 2:

                    m, n = pair

                    if m.distance < (
                        ratio_threshold * n.distance
                    ):

                        good_matches.append(m)

            # --------------------------------------------------
            # STEP 13: Draw Matched Keypoints
            # --------------------------------------------------

            matched_image = cv2.drawMatches(
                resized1,
                keypoints1,
                resized2,
                keypoints2,
                good_matches,
                None,
                flags=cv2.DrawMatchesFlags_NOT_DRAW_SINGLE_POINTS
            )

            # --------------------------------------------------
            # STEP 14: Display Complete Pipeline
            # --------------------------------------------------

            plt.figure(figsize=(16, 12))

            # Image 1
            plt.subplot(2, 3, 1)
            plt.imshow(rgb1)
            plt.title("Original Image 1")
            plt.axis("off")

            # Image 2
            plt.subplot(2, 3, 2)
            plt.imshow(rgb2)
            plt.title("Original Image 2")
            plt.axis("off")

            # Preprocessed image 1
            plt.subplot(2, 3, 3)
            plt.imshow(
                normalized1,
                cmap="gray"
            )
            plt.title("Preprocessed Image 1")
            plt.axis("off")

            # Preprocessed image 2
            plt.subplot(2, 3, 4)
            plt.imshow(
                normalized2,
                cmap="gray"
            )
            plt.title("Preprocessed Image 2")
            plt.axis("off")

            # Keypoints image 1
            keypoint_image1 = cv2.drawKeypoints(
                normalized1,
                keypoints1,
                None,
                flags=cv2.DRAW_MATCHES_FLAGS_DRAW_RICH_KEYPOINTS
            )

            plt.subplot(2, 3, 5)
            plt.imshow(
                keypoint_image1,
                cmap="gray"
            )
            plt.title("SIFT Keypoints - Image 1")
            plt.axis("off")

            # Matching result
            plt.subplot(2, 3, 6)
            plt.imshow(matched_image)
            plt.title("Matched Keypoints")
            plt.axis("off")

            plt.tight_layout()
            plt.show()

            # --------------------------------------------------
            # STEP 15: Display Matching Result Separately
            # --------------------------------------------------

            plt.figure(figsize=(18, 8))

            plt.imshow(matched_image)

            plt.title(
                "End-to-End SIFT Feature Matching"
            )

            plt.axis("off")

            plt.show()

            # --------------------------------------------------
            # STEP 16: Display Numerical Results
            # --------------------------------------------------

            print(
                "========== END-TO-END FEATURE PIPELINE =========="
            )

            print("\nImage 1:")
            print(
                "Original shape:",
                rgb1.shape
            )

            print(
                "Preprocessed shape:",
                normalized1.shape
            )

            print(
                "SIFT keypoints:",
                len(keypoints1)
            )

            print("\nImage 2:")
            print(
                "Original shape:",
                rgb2.shape
            )

            print(
                "Preprocessed shape:",
                normalized2.shape
            )

            print(
                "SIFT keypoints:",
                len(keypoints2)
            )

            print("\nMatching:")
            print(
                "KNN match pairs:",
                len(knn_matches)
            )

            print(
                "Ratio threshold:",
                ratio_threshold
            )

            print(
                "Good matches:",
                len(good_matches)
            )

            print(
                "\nEnd-to-end feature pipeline "
                "completed successfully."
            )
```
