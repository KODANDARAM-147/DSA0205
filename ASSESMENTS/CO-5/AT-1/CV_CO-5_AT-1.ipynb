```python
# ============================================
# Facial Expression Recognition using DeepFace
# Google Colab
# ============================================

from deepface import DeepFace
import cv2
from google.colab import files
from google.colab.patches import cv2_imshow

print("=======================================")
print(" FACIAL EXPRESSION RECOGNITION SYSTEM ")
print("=======================================\n")

print("Upload an image containing a human face.\n")

uploaded = files.upload()

image_path = list(uploaded.keys())[0]

print("\nAnalyzing Emotion...\n")

result = DeepFace.analyze(
    img_path=image_path,
    actions=['emotion'],
    enforce_detection=False
)

emotion = result[0]['dominant_emotion']

image = cv2.imread(image_path)

cv2.putText(
    image,
    "Emotion : " + emotion.upper(),
    (20,40),
    cv2.FONT_HERSHEY_SIMPLEX,
    1,
    (0,255,0),
    2
)

cv2_imshow(image)

print("\n==============================")
print("Dominant Emotion :", emotion.upper())
print("==============================\n")

print("Emotion Confidence:\n")

for key,value in result[0]["emotion"].items():
    print(f"{key:10s} : {value:.2f}%")
```

```python
# ==========================================================
# OCR Pipeline for Scanned Document Text Extraction
# Using OpenCV + Tesseract
# Google Colab
# ==========================================================

import cv2
import pytesseract
from PIL import Image
from google.colab import files
from google.colab.patches import cv2_imshow

print("==============================================")
print(" OCR PIPELINE FOR SCANNED DOCUMENT EXTRACTION ")
print("==============================================\n")

print("Upload a scanned document image (JPG/PNG).\n")

uploaded = files.upload()

image_path = list(uploaded.keys())[0]

# Read image
image = cv2.imread(image_path)

# Display original image
print("\nOriginal Image\n")
cv2_imshow(image)

# Convert to grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Noise removal
gray = cv2.GaussianBlur(gray, (3,3), 0)

# Thresholding
gray = cv2.threshold(
    gray,
    0,
    255,
    cv2.THRESH_BINARY + cv2.THRESH_OTSU
)[1]

print("\nProcessed Image\n")
cv2_imshow(gray)

# OCR
text = pytesseract.image_to_string(gray)

print("\n======================")
print(" EXTRACTED TEXT ")
print("======================\n")

print(text)

# Save extracted text
with open("Extracted_Text.txt","w",encoding="utf-8") as f:
    f.write(text)

print("\nText saved as Extracted_Text.txt")
```

```python
# ==========================================================
# Motion Detection System
# Google Colab Version
# ==========================================================

import cv2
from google.colab import files
from google.colab.patches import cv2_imshow

print("====================================")
print("VIDEO SURVEILLANCE MOTION DETECTOR")
print("====================================")

print("\nUpload a video file (.mp4 or .avi)\n")

uploaded = files.upload()

video_path = list(uploaded.keys())[0]

cap = cv2.VideoCapture(video_path)

ret, frame1 = cap.read()
ret, frame2 = cap.read()

frame_number = 0

while ret:

    frame_number += 1

    diff = cv2.absdiff(frame1, frame2)

    gray = cv2.cvtColor(diff, cv2.COLOR_BGR2GRAY)

    blur = cv2.GaussianBlur(gray, (5,5), 0)

    _, thresh = cv2.threshold(blur, 20, 255, cv2.THRESH_BINARY)

    dilated = cv2.dilate(thresh, None, iterations=3)

    contours, _ = cv2.findContours(
        dilated,
        cv2.RETR_TREE,
        cv2.CHAIN_APPROX_SIMPLE
    )

    motion = False

    for contour in contours:

        if cv2.contourArea(contour) < 1000:
            continue

        motion = True

        x,y,w,h = cv2.boundingRect(contour)

        cv2.rectangle(
            frame1,
            (x,y),
            (x+w,y+h),
            (0,255,0),
            2
        )

    if motion:
        cv2.putText(
            frame1,
            "MOTION DETECTED",
            (20,40),
            cv2.FONT_HERSHEY_SIMPLEX,
            1,
            (0,0,255),
            2
        )
    else:
        cv2.putText(
            frame1,
            "NO MOTION",
            (20,40),
            cv2.FONT_HERSHEY_SIMPLEX,
            1,
            (255,0,0),
            2
        )

    print("Frame:", frame_number)

    cv2_imshow(frame1)

    frame1 = frame2

    ret, frame2 = cap.read()

cap.release()

print("\nProcessing Completed.")
```

```python
import cv2
from google.colab import files
from google.colab.patches import cv2_imshow

print("======================================")
print(" REAL-TIME OBJECT TRACKING (Video)")
print("======================================")

print("\nUpload a video (.mp4 or .avi)\n")

uploaded = files.upload()

video_path = list(uploaded.keys())[0]

cap = cv2.VideoCapture(video_path)

success, frame = cap.read()

if not success:
    raise Exception("Unable to read video.")

frame = cv2.resize(frame, (640,480))

# Create CSRT Tracker
tracker = cv2.TrackerCSRT_create()

# ----------------------------------------------------
# CHANGE THESE VALUES TO MATCH YOUR OBJECT
# x, y = top-left corner
# w, h = width and height
# ----------------------------------------------------

bbox = (100,100,120,120)

tracker.init(frame, bbox)

frame_no = 1

while True:

    success, frame = cap.read()

    if not success:
        break

    frame = cv2.resize(frame,(640,480))

    ok, bbox = tracker.update(frame)

    if ok:

        x,y,w,h = [int(v) for v in bbox]

        cv2.rectangle(
            frame,
            (x,y),
            (x+w,y+h),
            (0,255,0),
            2
        )

        cv2.putText(
            frame,
            "Tracking",
            (20,40),
            cv2.FONT_HERSHEY_SIMPLEX,
            1,
            (0,255,0),
            2
        )

    else:

        cv2.putText(
            frame,
            "Lost",
            (20,40),
            cv2.FONT_HERSHEY_SIMPLEX,
            1,
            (0,0,255),
            2
        )

    print("Frame:", frame_no)

    cv2_imshow(frame)

    frame_no += 1

cap.release()

print("\nTracking Completed.")
```

```python
import cv2
from google.colab import files
from google.colab.patches import cv2_imshow

print("==========================================")
print(" FEATURE MATCHING FOR OBJECT RECOGNITION ")
print("==========================================")

print("\nUpload TWO images:")
print("1. Reference object image")
print("2. Scene image containing the object\n")

uploaded = files.upload()

file_names = list(uploaded.keys())

if len(file_names) != 2:
    raise Exception("Please upload exactly TWO images.")

img1 = cv2.imread(file_names[0], cv2.IMREAD_GRAYSCALE)
img2 = cv2.imread(file_names[1], cv2.IMREAD_GRAYSCALE)

orb = cv2.ORB_create(1000)

kp1, des1 = orb.detectAndCompute(img1, None)
kp2, des2 = orb.detectAndCompute(img2, None)

if des1 is None or des2 is None:
    raise Exception("Could not detect enough features in one or both images.")

bf = cv2.BFMatcher(cv2.NORM_HAMMING, crossCheck=True)

matches = bf.match(des1, des2)

matches = sorted(matches, key=lambda x: x.distance)

result = cv2.drawMatches(
    img1,
    kp1,
    img2,
    kp2,
    matches[:50],
    None,
    flags=2
)

print("\nNumber of Matches:", len(matches))

cv2_imshow(result)

cv2.imwrite("Feature_Matching_Result.jpg", result)

print("\nResult saved as Feature_Matching_Result.jpg")
```

```python
import cv2
from google.colab import files
from google.colab.patches import cv2_imshow

print("========================================")
print("OBJECT RECOGNITION USING FEATURE MATCHING")
print("========================================")

# Upload reference image
print("\nUpload Reference Image")
uploaded = files.upload()
reference_path = list(uploaded.keys())[0]

# Upload scene image
print("\nUpload Scene Image")
uploaded = files.upload()
scene_path = list(uploaded.keys())[0]

# Read images
img1 = cv2.imread(reference_path, cv2.IMREAD_GRAYSCALE)
img2 = cv2.imread(scene_path, cv2.IMREAD_GRAYSCALE)

# Check if images loaded
if img1 is None or img2 is None:
    raise Exception("Error reading one or both images.")

# Create ORB detector
orb = cv2.ORB_create(1000)

# Detect keypoints
kp1, des1 = orb.detectAndCompute(img1, None)
kp2, des2 = orb.detectAndCompute(img2, None)

if des1 is None or des2 is None:
    raise Exception("Not enough features detected in one of the images.")

# Matcher
bf = cv2.BFMatcher(cv2.NORM_HAMMING, crossCheck=True)

matches = bf.match(des1, des2)

matches = sorted(matches, key=lambda x: x.distance)

# Draw first 50 matches
result = cv2.drawMatches(
    img1,
    kp1,
    img2,
    kp2,
    matches[:50],
    None,
    flags=cv2.DrawMatchesFlags_NOT_DRAW_SINGLE_POINTS
)

print("Total Matches Found:", len(matches))

cv2_imshow(result)

cv2.imwrite("Feature_Matching_Result.jpg", result)

print("Result saved as Feature_Matching_Result.jpg")
```

```python
# ==========================================================
# Automated License Plate Region Localization
# Google Colab
# ==========================================================

import cv2
import numpy as np
from google.colab import files
from google.colab.patches import cv2_imshow

print("==========================================")
print(" LICENSE PLATE REGION LOCALIZATION ")
print("==========================================")

print("\nUpload a vehicle image (.jpg or .png)\n")

uploaded = files.upload()

image_path = list(uploaded.keys())[0]

# Read image
image = cv2.imread(image_path)

if image is None:
    raise Exception("Unable to read image.")

# Resize for better visualization
image = cv2.resize(image, (800, 600))

# Create a copy
output = image.copy()

# Convert to grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Reduce noise
gray = cv2.bilateralFilter(gray, 11, 17, 17)

# Detect edges
edged = cv2.Canny(gray, 30, 200)

# Find contours
contours, _ = cv2.findContours(
    edged.copy(),
    cv2.RETR_TREE,
    cv2.CHAIN_APPROX_SIMPLE
)

# Sort contours by area
contours = sorted(contours, key=cv2.contourArea, reverse=True)[:20]

plate = None

for contour in contours:

    perimeter = cv2.arcLength(contour, True)

    approx = cv2.approxPolyDP(
        contour,
        0.018 * perimeter,
        True
    )

    # License plate usually has 4 corners
    if len(approx) == 4:
        plate = approx
        break

if plate is not None:

    cv2.drawContours(
        output,
        [plate],
        -1,
        (0,255,0),
        3
    )

    x, y, w, h = cv2.boundingRect(plate)

    roi = image[y:y+h, x:x+w]

    print("\nLicense Plate Detected!\n")

    print("Plate Region")

    cv2_imshow(roi)

else:

    print("\nNo License Plate Found.")

print("\nVehicle Image")

cv2_imshow(output)

cv2.imwrite("License_Plate_Localized.jpg", output)

print("\nResult saved as License_Plate_Localized.jpg")
```

```python
# ==========================================================
# Background Subtraction Surveillance Pipeline
# Google Colab Version
# ==========================================================

import cv2
from google.colab import files
from google.colab.patches import cv2_imshow

print("==============================================")
print("BACKGROUND SUBTRACTION SURVEILLANCE PIPELINE")
print("==============================================")

print("\nUpload a surveillance video (.mp4 or .avi)\n")

uploaded = files.upload()

video_path = list(uploaded.keys())[0]

if not video_path.lower().endswith((".mp4", ".avi", ".mov", ".mkv")):
    raise Exception("Please upload a valid video file.")

cap = cv2.VideoCapture(video_path)

if not cap.isOpened():
    raise Exception("Unable to open video.")

# Background Subtractor
bg_subtractor = cv2.createBackgroundSubtractorMOG2(
    history=500,
    varThreshold=25,
    detectShadows=True
)

frame_number = 0

while True:

    ret, frame = cap.read()

    if not ret:
        break

    frame_number += 1

    frame = cv2.resize(frame, (640,480))

    # Apply background subtraction
    fg_mask = bg_subtractor.apply(frame)

    # Remove noise
    kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (3,3))
    fg_mask = cv2.morphologyEx(
        fg_mask,
        cv2.MORPH_OPEN,
        kernel
    )

    contours, _ = cv2.findContours(
        fg_mask,
        cv2.RETR_EXTERNAL,
        cv2.CHAIN_APPROX_SIMPLE
    )

    for contour in contours:

        if cv2.contourArea(contour) < 500:
            continue

        x, y, w, h = cv2.boundingRect(contour)

        cv2.rectangle(
            frame,
            (x, y),
            (x+w, y+h),
            (0,255,0),
            2
        )

        cv2.putText(
            frame,
            "Moving Object",
            (x, y-10),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.6,
            (0,255,0),
            2
        )

    print("Frame:", frame_number)

    print("Foreground Mask")
    cv2_imshow(fg_mask)

    print("Detected Objects")
    cv2_imshow(frame)

cap.release()

print("\nProcessing Completed Successfully.")
```

```python
# ==========================================================
# Multi-Face Detection and Counting from Video
# Google Colab Version
# ==========================================================

import cv2
import os
from google.colab import files
from google.colab.patches import cv2_imshow

print("===========================================")
print("MULTI-FACE DETECTION AND COUNTING PIPELINE")
print("===========================================\n")

print("Upload a video (.mp4, .avi, .mov)\n")

uploaded = files.upload()

video_path = list(uploaded.keys())[0]

# Check file
if not os.path.exists(video_path):
    raise Exception("Video file not found.")

# Open video
cap = cv2.VideoCapture(video_path)

if not cap.isOpened():
    raise Exception("Unable to open video. Try another MP4 file.")

# Load Haar Cascade
face_cascade = cv2.CascadeClassifier(
    cv2.data.haarcascades +
    "haarcascade_frontalface_default.xml"
)

if face_cascade.empty():
    raise Exception("Failed to load Haar Cascade.")

frame_number = 0

while True:

    ret, frame = cap.read()

    if not ret:
        break

    frame_number += 1

    frame = cv2.resize(frame, (640,480))

    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    faces = face_cascade.detectMultiScale(
        gray,
        scaleFactor=1.1,
        minNeighbors=5,
        minSize=(30,30)
    )

    # Draw rectangles
    for (x,y,w,h) in faces:

        cv2.rectangle(
            frame,
            (x,y),
            (x+w,y+h),
            (0,255,0),
            2
        )

    # Display count
    cv2.putText(
        frame,
        f"Faces: {len(faces)}",
        (20,40),
        cv2.FONT_HERSHEY_SIMPLEX,
        1,
        (0,0,255),
        2
    )

    print(f"Frame {frame_number} : {len(faces)} face(s)")

    cv2_imshow(frame)

cap.release()

print("\nProcessing Completed Successfully.")
```

```python
# ==========================================================
# Complete Image-to-Video Processing Pipeline
# Google Colab Version
# ==========================================================

import cv2
import os
from google.colab import files
from google.colab.patches import cv2_imshow

print("===============================================")
print(" COMPLETE IMAGE TO VIDEO PROCESSING PIPELINE ")
print("===============================================\n")

print("Upload multiple images (.jpg/.png)")
print("Select at least 3 images.\n")

uploaded = files.upload()

image_files = list(uploaded.keys())

if len(image_files) < 3:
    raise Exception("Please upload at least 3 images.")

# Read first image
first = cv2.imread(image_files[0])

if first is None:
    raise Exception("Unable to read first image.")

height, width = first.shape[:2]

output_video = "Processed_Video.mp4"

fourcc = cv2.VideoWriter_fourcc(*'mp4v')

writer = cv2.VideoWriter(
    output_video,
    fourcc,
    2,          # Frames Per Second
    (width, height)
)

frame_number = 1

for img_name in image_files:

    image = cv2.imread(img_name)

    if image is None:
        continue

    image = cv2.resize(image, (width, height))

    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

    edges = cv2.Canny(gray, 80, 150)

    processed = cv2.cvtColor(edges, cv2.COLOR_GRAY2BGR)

    cv2.putText(
        processed,
        f"Frame {frame_number}",
        (20,40),
        cv2.FONT_HERSHEY_SIMPLEX,
        1,
        (0,255,0),
        2
    )

    print(f"Processing Frame {frame_number}")

    cv2_imshow(processed)

    writer.write(processed)

    frame_number += 1

writer.release()

print("\n====================================")
print("VIDEO CREATED SUCCESSFULLY")
print("====================================")

print("Output File :", output_video)

files.download(output_video)
```
