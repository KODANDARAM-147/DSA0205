```python
import cv2
from google.colab import files
from google.colab.patches import cv2_imshow

# Upload an image
uploaded = files.upload()

# Get the uploaded file name
image_name = list(uploaded.keys())[0]

# Read the uploaded image
img = cv2.imread(image_name)

# Check if the image is loaded successfully
if img is None:
    print("Error: Could not load the image.")
else:
    # Display the original image
    cv2_imshow(img)

    # Scale the image to bigger size (2x)
    bigger = cv2.resize(img, None, fx=2, fy=2, interpolation=cv2.INTER_LINEAR)

    # Scale the image to smaller size (0.5x)
    smaller = cv2.resize(img, None, fx=0.5, fy=0.5, interpolation=cv2.INTER_AREA)

    # Display the bigger image
    print("Bigger Image")
    cv2_imshow(bigger)

    # Display the smaller image
    print("Smaller Image")
    cv2_imshow(smaller)
```
