```python
import cv2
from google.colab import files
from google.colab.patches import cv2_imshow

# Upload image
uploaded = files.upload()
image_name = list(uploaded.keys())[0]

# Read image
img = cv2.imread(image_name)

if img is None:
    print("Error: Could not load the image.")
else:
    print("Original Image")
    cv2_imshow(img)

    # Clockwise Rotation (90°)
    clockwise = cv2.rotate(img, cv2.ROTATE_90_CLOCKWISE)

    # Counter Clockwise Rotation (90°)
    counter_clockwise = cv2.rotate(img, cv2.ROTATE_90_COUNTERCLOCKWISE)

    print("Clockwise Rotated Image")
    cv2_imshow(clockwise)

    print("Counter Clockwise Rotated Image")
    cv2_imshow(counter_clockwise)
```
