```python
import cv2
from google.colab import files
from IPython.display import HTML, display
from base64 import b64encode

# Upload video
uploaded = files.upload()
video_name = list(uploaded.keys())[0]

# Read uploaded video
cap = cv2.VideoCapture(video_name)

if not cap.isOpened():
    print("Error: Could not open video.")
else:
    width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
    height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
    fps = cap.get(cv2.CAP_PROP_FPS)

    # Output video writers
    fourcc = cv2.VideoWriter_fourcc(*'mp4v')

    slow_writer = cv2.VideoWriter("slow_motion.mp4", fourcc, max(fps/2, 1), (width, height))
    fast_writer = cv2.VideoWriter("fast_motion.mp4", fourcc, fps*2, (width, height))

    while True:
        ret, frame = cap.read()
        if not ret:
            break

        slow_writer.write(frame)
        fast_writer.write(frame)

    cap.release()
    slow_writer.release()
    fast_writer.release()

    print("Slow Motion Video:")

    with open("slow_motion.mp4", "rb") as f:
        slow_data = f.read()

    slow_url = "data:video/mp4;base64," + b64encode(slow_data).decode()

    display(HTML(f"""
    <video width="500" controls>
      <source src="{slow_url}" type="video/mp4">
    </video>
    """))

    print("Fast Motion Video:")

    with open("fast_motion.mp4", "rb") as f:
        fast_data = f.read()

    fast_url = "data:video/mp4;base64," + b64encode(fast_data).decode()

    display(HTML(f"""
    <video width="500" controls>
      <source src="{fast_url}" type="video/mp4">
    </video>
    """))

    # Download videos
    files.download("slow_motion.mp4")
    files.download("fast_motion.mp4")
```
