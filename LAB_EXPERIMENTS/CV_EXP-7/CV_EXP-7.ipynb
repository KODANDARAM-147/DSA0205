```python
import cv2
from google.colab import files

# Upload the recorded video
uploaded = files.upload()
video_name = list(uploaded.keys())[0]

cap = cv2.VideoCapture(video_name)

# Get video properties
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

# Assume original video FPS is 30
fps = 30

# Create output videos
fourcc = cv2.VideoWriter_fourcc(*'mp4v')

# Slow Motion (plays at 15 FPS)
slow = cv2.VideoWriter("slow_motion.mp4", fourcc, 15, (width, height))

# Fast Motion (plays at 60 FPS)
fast = cv2.VideoWriter("fast_motion.mp4", fourcc, 60, (width, height))

while True:
    ret, frame = cap.read()
    if not ret:
        break

    # Slow motion: write every frame
    slow.write(frame)

    # Fast motion: skip every alternate frame
    ret2, frame2 = cap.read()
    if ret2:
        fast.write(frame2)

cap.release()
slow.release()
fast.release()

print("Processing Completed Successfully!")

# Download output videos
files.download("slow_motion.mp4")
files.download("fast_motion.mp4")
```
